## 6. Gierig und nicht gierig (optional)

Wiederholungen wie `*` sind standardmäßig **gierig**: Sie versuchen zunächst, möglichst viel Text zu erfassen. Mit `*?` versucht die Wiederholung zunächst, möglichst wenig Text zu erfassen, sodass der restliche Ausdruck noch passen kann.

Führe die folgende Zelle aus und vergleiche die Treffer. Was ändert sich zwischen den beiden Mustern?

Dokumentation: [Wiederholungen in regulären Ausdrücken](https://docs.python.org/3.13/library/re.html#regular-expression-syntax).

```python
label_text = 'Labels: <steel> and <wood>.'

print('Greedy:', re.findall(r'<.*>', label_text))
print('Non-greedy:', re.findall(r'<.*?>', label_text))
```
