# Reguläre Ausdrücke: Python-API und Feldprüfung

```python
import re
```

## 1. Den ersten Ziffernblock finden

Formuliere `pattern_01` für eine Folge aus einer oder mehreren Ziffern. Implementiere `first_number`: Gib den ersten Ziffernblock in `text` als String zurück. Gibt es keinen Treffer, gib `None` zurück.

Verwende `re.search` und prüfe, ob ein Treffer vorliegt, bevor du ihn ausliest.

### Python-API: `re.search` und `Match.group`

`re.search(pattern, string, flags=0)` sucht im gesamten `string` nach dem ersten Treffer für `pattern`. `flags` ist optional und verändert bei Bedarf die Suche. Bei einem Treffer gibt die Funktion ein Match-Objekt zurück, sonst `None`.

`match.group(0)` liefert den gesamten gefundenen Text. Du darfst `group` erst aufrufen, wenn tatsächlich ein Match-Objekt vorliegt.

Dokumentation: [`re.search`](https://docs.python.org/3.13/library/re.html#re.search), [`Match.group`](https://docs.python.org/3.13/library/re.html#re.Match.group).

```python
text_01 = ('The delivery reference is 6841 and the package is stored in aisle 9.')

pattern_01 = r'[0-9]+'
```

```python
def first_number(text: str, pattern: str) -> str | None:
    match = re.search(pattern, text)
    if not match:
        return None
    return match.group(0)
    raise NotImplementedError('Implement first_number')
```

```python
assert first_number(text_01, pattern_01) == '6841'
assert first_number('The archives are on level 7 and floor 12.', pattern_01) == '7'
assert first_number('This report contains no numerical identifier.', pattern_01) is None
assert first_number('', pattern_01) is None
assert first_number('A cat rests nearby.', r'cat') == 'cat'

print('Tests passed.')
```

## 2. Einen Wert vollständig prüfen

Eine PIN soll genau vier Ziffern enthalten. Formuliere `pattern_02` und implementiere `is_four_digit_value`: Gib `True` zurück, wenn der gesamte Wert in `text` passt, sonst `False`. Verwende `re.fullmatch`.

### Python-API: `re.fullmatch`

`re.fullmatch(pattern, string, flags=0)` prüft, ob `pattern` auf den gesamten `string` passt. `flags` ist optional. Die Funktion gibt ein Match-Objekt zurück, wenn der gesamte Wert passt, andernfalls `None`.

Anders als bei `re.search` reicht ein passender Ausschnitt innerhalb des Werts nicht aus. Die Funktion prüft hier nur die Zeichenform; ob die PIN tatsächlich vergeben wurde, bleibt offen.

Dokumentation: [`re.fullmatch`](https://docs.python.org/3.13/library/re.html#re.fullmatch), [Vergleich von Suchfunktionen](https://docs.python.org/3.13/library/re.html#search-vs-match).

```python
text_to_match_02 = '4816'
text_not_to_match_02 = 'My PIN number is 4816, buddy.'

pattern_02 = r'[0-9]{4}'
```

```python
def is_four_digit_value(text: str, pattern: str) -> bool:
    match = re.fullmatch(pattern, text)
    if not match:
        return False
    return True
    raise NotImplementedError('Implement is_four_digit_value')
```

```python
assert is_four_digit_value(text_to_match_02, pattern_02) is True
assert is_four_digit_value('0000', pattern_02) is True
assert is_four_digit_value(text_not_to_match_02, pattern_02) is False
assert is_four_digit_value('48163', pattern_02) is False
assert is_four_digit_value('48A6', pattern_02) is False
assert is_four_digit_value(' 4816', pattern_02) is False
assert is_four_digit_value('', pattern_02) is False
assert is_four_digit_value('cat', r'cat') is True

print('Tests passed.')
```

## 3. Mehrere Treffer und Gruppen auslesen

Finde Messwerte aus einer ganzen Zahl, einem Leerzeichen und der Einheit `ml` oder `l`. Formuliere `pattern_03` mit zwei Gruppen: eine für die Zahl und eine für die Einheit.

Implementiere `find_measurements`. Gib mit `re.findall` eine Liste von Tupeln aus Zahl und Einheit zurück.

### Python-API: `re.findall`

`re.findall(pattern, string, flags=0)` sucht alle nicht überlappenden Treffer im `string` und gibt sie als Liste zurück. `flags` ist optional. Ohne Capture Group enthält die Liste die gesamten Treffer. Bei einer Gruppe enthält sie deren Inhalte; bei mehreren Gruppen enthält sie Tupel mit den jeweiligen Gruppeninhalten.

Dokumentation: [`re.findall`](https://docs.python.org/3.13/library/re.html#re.findall).

```python
text_03 = (
    'The laboratory prepared samples for the test. '
    'The first container holds 250 ml of water. '
    'The second container holds 3 l of solution.'
)

pattern_03 = r'([0-9]+) (l|ml)'
```

```python
def find_measurements(text: str, pattern: str) -> list[tuple[str, str]]:
    return re.findall(pattern, text)
    
    raise NotImplementedError('Implement find_measurements')
```

```python
assert find_measurements(text_03, pattern_03) == [('250', 'ml'), ('3', 'l')]
assert find_measurements('The sample contains 75 ml.', pattern_03) == [('75', 'ml')]
assert find_measurements('There are no measurements.', pattern_03) == []
assert find_measurements('cat-dog', r'(cat)-(dog)') == [('cat', 'dog')]

print('Tests passed.')
```

## 4. Datumsangaben auslesen

Finde Angaben, bei denen auf `Inspection date: ` ein Datum in der Form `TT.MM.JJJJ` folgt: zwei Ziffern für den Tag, ein Punkt, zwei für den Monat, ein Punkt und vier für das Jahr. Nach dem Datum darf weiterer Text stehen.

Formuliere `pattern_04` und erfasse nur das Datum in einer Gruppe. Implementiere `find_dates` und gib mit `re.findall` eine Liste der gefundenen Datumsangaben zurück.

Die Suche findet passende Ausschnitte. Sie prüft weder den gesamten Feldwert noch, ob ein Datum im Kalender existiert.

Dokumentation: [`re.findall`](https://docs.python.org/3.13/library/re.html#re.findall).


```python
text_04 = (
    'Product: Example floor panel\n'
    'The first report notes Inspection date: 04.03.2025 in the record.\n'
    'Inspection date: 21.09.2026\n'
    'Inspection date: 21.09.26\n'
    'Inspection date: 21.0X.2026\n'
)

pattern_04 = r'Inspection date: ((?:\d{2}\.){2}\d{4})'
```

```python
def find_dates(text: str, pattern: str) -> list[str]:
    return re.findall(pattern, text)
    raise NotImplementedError('Implement find_dates')
```

```python
assert find_dates(text_04, pattern_04) == ['04.03.2025', '21.09.2026']
assert find_dates('Inspection date: 21.09.26', pattern_04) == []
assert find_dates('No inspection date is recorded.', pattern_04) == []
assert find_dates('pet: cat', r'pet: (cat)') == ['cat']

print('Tests passed.')
```

## 5. Datumswerte vollständig prüfen

Ein Datumswert soll genau die Form `TT.MM.JJJJ` haben. Formuliere `pattern_05` und implementiere `is_date_format`: Gib `True` zurück, wenn der gesamte Wert in `text` passt, sonst `False`. Verwende `re.fullmatch`.

Die Prüfung betrifft nur die Schreibform, nicht die Gültigkeit im Kalender.

Dokumentation: [`re.fullmatch`](https://docs.python.org/3.13/library/re.html#re.fullmatch).


```python
text_05 = '04.03.2025'

pattern_05 = r'(?:\d{2}\.){2}\d{4}'
```

```python
def is_date_format(text: str, pattern: str) -> bool:
    match = re.fullmatch(pattern, text)
    return match.group() if match else None
    raise NotImplementedError('Implement is_date_format')
```

```python
assert is_date_format(text_05, pattern_05) is True
assert is_date_format('21.09.26', pattern_05) is False
assert is_date_format('21.0X.2026', pattern_05) is False
assert is_date_format('21.09.20267', pattern_05) is False
assert is_date_format('', pattern_05) is False
assert is_date_format('cat', r'cat') is True

print('Tests passed.')
```

## 6. Gierig und nicht gierig (optional)

Wiederholungen wie `*` sind standardmäßig **gierig**: Sie versuchen zunächst, möglichst viel Text zu erfassen. Mit `*?` versucht die Wiederholung zunächst, möglichst wenig Text zu erfassen, sodass der restliche Ausdruck noch passen kann.

Führe die folgende Zelle aus und vergleiche die Treffer. Was ändert sich zwischen den beiden Mustern?

Dokumentation: [Wiederholungen in regulären Ausdrücken](https://docs.python.org/3.13/library/re.html#regular-expression-syntax).

```python
label_text = 'Labels: <steel> and <wood>.'

print('Greedy:', re.findall(r'<.*>', label_text))
print('Non-greedy:', re.findall(r'<.*?>', label_text))
```
