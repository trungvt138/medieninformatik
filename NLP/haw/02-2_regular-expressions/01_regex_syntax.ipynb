# Reguläre Ausdrücke: Muster anwenden

**Leitfrage:** Wie lassen sich Zeichenfolgen mit regulären Ausdrücken finden?

## Arbeitsauftrag

Führe die folgende Zelle aus. Ergänze anschließend bei jeder Aufgabe das Muster und führe die zugehörige Testzelle aus. Die Aufgaben 8–10 sind eine Vertiefung.

Verwende für Muster Raw Strings wie `r'...'`. Zum Ausprobieren kannst du [regex101](https://regex101.com/) mit dem Python-Flavor verwenden; gib dort das Muster ohne `r` und Anführungszeichen ein.

Mit Ziffern sind in diesen Aufgaben die Zeichen `0` bis `9` gemeint.

```python
import re


def check_matches(pattern: str, text: str, expected: list[tuple[str, int, int]]) -> None:
    actual = [(match.group(), *match.span()) for match in re.finditer(pattern, text)]
    print('Matches:', actual)
    assert actual == expected, f'Expected {expected}, got {actual}'
    print('Tests passed.')
```

## 1. Literale

Finde jedes kleine `c`. Ein großes `C` soll nicht als Treffer zählen.

```python
text_01 = 'A cat rests near the bench while the dog sleeps in the afternoon.'

pattern_01 = 'c'
```

```python
check_matches(pattern_01, text_01, [('c', 2, 3), ('c', 24, 25)])
```

## 2. Einen Punkt suchen

Finde alle tatsächlichen Punkte.

```python
text_02 = 'Contact a.b@example.com. Please read section 3.1 before tomorrow.'

pattern_02 = r'\.'
```

```python
check_matches(pattern_02, text_02, [('.', 9, 10), ('.', 19, 20), ('.', 23, 24), ('.', 46, 47), ('.', 64, 65)])
```

## 3. Zeichenklasse

Finde alle Ziffern als einzelne Treffer.

```python
text_03 = 'The package is in aisle 2, shelf 7, next to the open window.'

pattern_03 = r'\d'
```

```python
check_matches(pattern_03, text_03, [('2', 24, 25), ('7', 33, 34)])
```

## 4. Negierte Zeichenklasse

Finde alle einzelnen Zeichen, die keine Ziffern sind.

```python
text_04 = '9780140328721-9780261103344/9780061120084'

pattern_04 = r'\D'
```

```python
check_matches(pattern_04, text_04, [('-', 13, 14), ('/', 27, 28)])
```

## 5. Alternativen

Finde die Wörter `cat` und `dog`.

```python
text_05 = 'The cat watches the garden, while a dog waits beside the door.'

pattern_05 = r'(cat|dog)'
```

```python
check_matches(pattern_05, text_05, [('cat', 4, 7), ('dog', 36, 39)])
```

## 6. Wiederholung

Finde jede zusammenhängende Folge aus einer oder mehreren Ziffern. `12` soll ein einzelner Treffer sein.

```python
text_06 = 'The first shelf holds 12 boxes; the second holds 3 boxes.'

pattern_06 = r'[1-9]*[0-9]'
```

```python
check_matches(pattern_06, text_06, [('12', 22, 24), ('3', 49, 50)])
```

## 7. Leerraum

Finde alle Leerraumzeichen. Der Text enthält Tabulatoren zwischen den Feldern.

```python
text_07 = 'shipment_approved\t12kg\tready_for_delivery'

pattern_07 = r'\t'
```

```python
check_matches(pattern_07, text_07, [('\t', 17, 18), ('\t', 22, 23)])
```

## 8. Wortgrenzen (Vertiefung)

Finde das eigenständige Wort `cat`, nicht den Teil in `concatenate`.

```python
text_08 = 'The cat sleeps by the door; concatenate is a command; another cat waits outside.'

pattern_08 = r'\bcat\b'
```

```python
check_matches(pattern_08, text_08, [('cat', 4, 7), ('cat', 62, 65)])
```

## 9. Fünfstellige Werte (Vertiefung)

Finde eigenständige Folgen aus genau fünf Ziffern. Die sechsstellige Referenznummer soll kein Treffer sein.

```python
text_09 = 'Delivery addresses: 20095 Hamburg and 10115 Berlin; reference 200950 and code 20A95 are invalid.'

pattern_09 = r'\b[0-9]{5}\b'
```

```python
check_matches(pattern_09, text_09, [('20095', 20, 25), ('10115', 38, 43)])
```

## 10. Messwerte (Vertiefung)

Finde Messwerte nach dieser Regel: mindestens eine Ziffer, optional ein Komma mit mindestens einer weiteren Ziffer, optional genau ein Leerzeichen und anschließend `kg` oder `g`. Angaben wie `12, kg` und `12 mg` sollen nicht passen.

```python
text_10 = 'The shipment weighs 12,5 kg; each sample weighs 7 g; another box weighs 12kg. Ignore 12, kg and 12 mg.'

pattern_10 = r'[0-9]+(\,[0-9]+)? ?(kg|g)'
```

```python
check_matches(pattern_10, text_10, [('12,5 kg', 20, 27), ('7 g', 48, 51), ('12kg', 72, 76)])
```
