## 2 Hamming-Distanz


Die Hamming-Distanz ist die **absolute Zahl** verschiedener Positionen in zwei gleich langen Strings:

$$d_H(x,y)=\sum_{i=1}^{n}\mathbf{1}[x_i\ne y_i],\qquad |x|=|y|=n.$$

$\mathbf{1}[P]$ ist 1, wenn $P$ zutrifft, sonst 0. Bei unterschiedlichen Längen soll die Funktion `ValueError` auslösen.


```python
def hamming_distance(w1:str, w2:str) -> int:
    if len(w1) != len(w2):
        raise ValueError("Strings must have the same length")

    result = 0

    if len(w1) != 0:
        for c1, c2 in zip(w1, w2):
            if c1 != c2:
                result += 1

    return result
    raise NotImplementedError('Implement hamming_distance')

```

### Testfälle


```python
import unittest
check = unittest.TestCase()

# assertEqual(actual, expected)
check.assertEqual(hamming_distance('', ''), 0)
check.assertEqual(hamming_distance('cat', 'cat'), 0)
check.assertEqual(hamming_distance('cat', 'car'), 1)
check.assertEqual(hamming_distance('cat', 'act'), 2)
check.assertEqual(hamming_distance('abc', 'xyz'), 3)
check.assertEqual(hamming_distance('Cat', 'cat'), 1)

# Hamming distance is only defined for strings of equal length.
with check.assertRaises(ValueError):
    hamming_distance('cat', 'coast')

with check.assertRaises(ValueError):
    hamming_distance('cart', 'cat')

print('Hamming tests passed.')
```

## 3 Jaccard-Ähnlichkeit


Für die Zeichenketten $x$ und $y$ ist die Jaccard-Ähnlichkeit ihrer **Zeichenmengen** $A=\operatorname{set}(x)$ und $B=\operatorname{set}(y)$ wie folgt definiert:

$$s_J(x,y)=\frac{|A\cap B|}{|A\cup B|}$$

Gleiche Mengen haben die Ähnlichkeit 1, disjunkte Mengen die Ähnlichkeit 0. Für zwei leere Mengen legen wir die Ähnlichkeit 1 fest.  
Da die Jaccard-Ähnlichkeit mit Mengen arbeitet, sind Reihenfolge und Häufigkeiten der Zeichen nicht relevant. 


```python
def jaccard_similarity(w1: str, w2: str) -> float:
    if len(w1) == 0 and len(w2) == 0 and w1 == w2:
        return 1.0
    s1, s2 = set(w1), set(w2)
    return len(s1 & s2)/len(s1 | s2)
    
    
    raise NotImplementedError('Implement jaccard_similarity')
```

### Testfälle


```python
import unittest
check = unittest.TestCase()

# assertEqual(actual, expected)
check.assertEqual(jaccard_similarity('', ''), 1.0)
check.assertEqual(jaccard_similarity('cat', ''), 0.0)
check.assertEqual(jaccard_similarity('cat', 'cat'), 1.0)
check.assertEqual(jaccard_similarity('cat', 'act'), 1.0)
check.assertEqual(jaccard_similarity('cat', 'dog'), 0.0)

# Compare floating-point results with a tolerance.
check.assertAlmostEqual(jaccard_similarity('cat', 'car'), 0.5, delta=1e-9)
check.assertAlmostEqual(jaccard_similarity('cart', 'coast'), 3 / 6, delta=1e-9)

print('Jaccard similarity tests passed.')
```

## 4 Längster gemeinsamer Substring


Ein **Substring** ist eine zusammenhängende Teilzeichenkette. Für zwei
Zeichenketten $x$ und $y$ suchen wir einen gemeinsamen Substring $s$ maximaler Länge.

Beispielsweise ist `ca` der längste gemeinsame Substring von `cat` und
`car`. Gibt es mehrere gleich lange längste Substrings, kann die Funktion
einen davon zurückgeben. Haben die Zeichenketten kein gemeinsames Zeichen,
ist das Ergebnis die leere Zeichenkette.

Eine Ähnlichkeit lässt sich definieren, indem wir die Länge von $s$ durch
die Länge der längeren Zeichenkette teilen:

$$
\operatorname{sim}(x, y) = \frac{|s|}{\max(|x|, |y|)}
$$

Für zwei leere Zeichenketten legen wir die Ähnlichkeit auf 1 fest.


```python
def longest_common_substring(w1: str, w2: str) -> str:

    if len(w1) <= len(w2):
        shorter, longer = w1, w2
    else:
        shorter, longer = w2, w1

    substrings = set()
    for start in range(len(shorter)):
        for end in range(start + 1, len(shorter) + 1):
            substrings.add(shorter[start:end])

    for substring in sorted(substrings, key=len, reverse=True):
        if substring in longer:
            return substring

    return ''


def lcs_similarity(w1: str, w2: str) -> float:
    
    if not w1 and not w2:
        return 1.0
    
    return len(longest_common_substring(w1, w2)) / max(len(w1), len(w2))
```

### Testfälle

```python
import unittest
check = unittest.TestCase()

check.assertEqual(longest_common_substring('', ''), '')
check.assertEqual(longest_common_substring('cat', ''), '')
check.assertEqual(longest_common_substring('', 'cat'), '')
check.assertEqual(longest_common_substring('cat', 'cat'), 'cat')
check.assertEqual(longest_common_substring('cat', 'dog'), '')
check.assertEqual(longest_common_substring('cat', 'car'), 'ca')
check.assertEqual(longest_common_substring('car', 'cat'), 'ca')
check.assertEqual(longest_common_substring('Cat', 'cat'), 'at')
check.assertEqual(longest_common_substring('intention', 'execution'), 'tion')

print('Longest common substring tests passed.')

check.assertAlmostEqual(lcs_similarity('', ''), 1.0)
check.assertAlmostEqual(lcs_similarity('cat', ''), 0.0)
check.assertAlmostEqual(lcs_similarity('', 'cat'), 0.0)
check.assertAlmostEqual(lcs_similarity('cat', 'cat'), 1.0)
check.assertAlmostEqual(lcs_similarity('cat', 'dog'), 0.0)
check.assertAlmostEqual(lcs_similarity('cat', 'car'), 2 / 3)
check.assertAlmostEqual(lcs_similarity('car', 'cat'), 2 / 3)
check.assertAlmostEqual(lcs_similarity('cats', 'scats'), 4 / 5)
check.assertAlmostEqual(lcs_similarity('intention', 'execution'), 4 / 9)

print('LCS similarity tests passed.')
```

```python
check.assertEqual(longest_common_substring('abcdef', 'abcXef'), 'abc')
check.assertAlmostEqual(lcs_similarity('abcdef', 'abcXef'), 3 / 6)

print('Only 1 of 6 characters differ between "abcdef" and "abcXef" but lcs_similarity is only 0.5')
```

## 5 Längste gemeinsame Teilsequenz


Eine **Teilsequenz** entsteht, indem Zeichen einer Zeichenkette weggelassen
werden. Die verbleibenden Zeichen müssen ihre Reihenfolge beibehalten,
aber nicht zusammenhängen.

Für zwei Zeichenketten $x$ und $y$ suchen wir eine gemeinsame Teilsequenz
$s$ maximaler Länge. Beispielsweise ist `abcef` eine längste gemeinsame
Teilsequenz von `abcdef` und `abcXef`.

Wie beim Substring können wir daraus eine Ähnlichkeit definieren:

$$
\operatorname{sim}(x, y) = \frac{|s|}{\max(|x|, |y|)}
$$

Für zwei leere Zeichenketten legen wir die Ähnlichkeit auf 1 fest. Gibt es
mehrere gleich lange längste Teilsequenzen, kann die Funktion eine davon
zurückgeben.
    

```python
def longest_common_subsequence(w1: str, w2: str) -> str:
    
    subsequences = [[''] * (len(w2) + 1) for _ in range(len(w1) + 1)]

    for i in range(1, len(w1) + 1):
        for j in range(1, len(w2) + 1):
            if w1[i - 1] == w2[j - 1]:
                subsequences[i][j] = subsequences[i - 1][j - 1] + w1[i - 1]
            else:
                subsequences[i][j] = max(
                    subsequences[i - 1][j],
                    subsequences[i][j - 1],
                    key=len,
                )

    return subsequences[-1][-1]


def subsequence_similarity(w1: str, w2: str) -> float:

    if not w1 and not w2:
        return 1.0

    return len(longest_common_subsequence(w1, w2)) / max(len(w1), len(w2))
    
```

```python
import unittest
check = unittest.TestCase()

check.assertEqual(longest_common_subsequence('', ''), '')
check.assertEqual(longest_common_subsequence('cat', ''), '')
check.assertEqual(longest_common_subsequence('', 'cat'), '')
check.assertEqual(longest_common_subsequence('cat', 'cat'), 'cat')
check.assertEqual(longest_common_subsequence('cat', 'dog'), '')
check.assertEqual(longest_common_subsequence('cat', 'car'), 'ca')
check.assertIn(longest_common_subsequence('cat', 'act'), {'at', 'ct'})

check.assertEqual(longest_common_subsequence('abcdef', 'abcXef'), 'abcef')
check.assertEqual(longest_common_subsequence('abcdef', 'aXbcYef'), 'abcef')
check.assertEqual(longest_common_subsequence('abcdef', 'aXbcYZef'), 'abcef')


print('Longest common subsequence tests passed.')

check.assertAlmostEqual(subsequence_similarity('', ''), 1.0)
check.assertAlmostEqual(subsequence_similarity('cat', ''), 0.0)
check.assertAlmostEqual(subsequence_similarity('cat', 'cat'), 1.0)
check.assertAlmostEqual(subsequence_similarity('cat', 'dog'), 0.0)
check.assertAlmostEqual(subsequence_similarity('cat', 'act'), 2 / 3)

check.assertAlmostEqual(subsequence_similarity('abcdef', 'abcXef'), 5 / 6)
check.assertAlmostEqual(subsequence_similarity('abcdef', 'aXbcYef'), 5 / 7)
check.assertAlmostEqual(subsequence_similarity('abcdef', 'aXbcYZef'), 5 / 8)

print('Subsequence similarity tests passed.')
```

```python
# Cases with suboptimal interpretability

check.assertAlmostEqual(subsequence_similarity('abcdef', 'fedcba'), 1 / 6)
check.assertAlmostEqual(subsequence_similarity('abcdef', 'ffffff'), 1 / 6)

check.assertAlmostEqual(subsequence_similarity('ab', 'abababab'), 2 / 8)
check.assertAlmostEqual(subsequence_similarity('abcdef', 'abababab'), 2 / 8)
```

## 6 Levenshtein-Distanz

Die [Teilsequenzähnlichkeit in Abschnitt 5](#5-Längste-gemeinsame-Teilsequenz) betrachtet,
wie viele Zeichen in derselben Reihenfolge erhalten bleiben.
Für eine Rechtschreibkorrektur ist es aber hilfreicher zu wissen,
wie viele Änderungen nötig sind, um ein Wort in ein anderes umzuwandeln.
Solche Maße werden als **Editierdistanzen** bezeichnet.

Die **Levenshtein-Distanz** ist die wohl bekannteste Editierdistanz.
Die Operationen **Einfügen, Löschen und Ersetzen** haben jeweils Kosten von 1.
Die Distanz ist die kleinstmögliche Anzahl dieser Operationen,
um die erste Zeichenkette in die zweite Zeichenkette umzuwandeln.

Beispiele:

- `cat` → `cats`: `s` einfügen → Distanz 1
- `cats` → `cat`: `s` löschen → Distanz 1
- `cat` → `cut`: `a` durch `u` ersetzen → Distanz 1
- `cat` → `actor`: `a` löschen, `a` einfügen, `o` einfügen, `r` einfügen → Distanz 4
- `cat` → `dog`: alle drei Zeichen ersetzen → Distanz 3


### 6.1 Die Matrix aufbauen

Wir berechnen die Levenshtein-Distanz mithilfe einer Distanzmatrix $D$.
Dafür betrachten wir die **Präfixe** der beiden Zeichenketten.
Ein Präfix besteht aus den ersten $k$ Zeichen einer Zeichenkette,
wobei $k$ zwischen 0 und der Länge der Zeichenkette liegt.
Das Präfix der Länge 0 ist die leere Zeichenkette `ε`.

Wir nennen die erste Zeichenkette **Quelle** und die zweite **Ziel**.
Die Zelle `D[i][j]` enthält die minimale Anzahl von Operationen,
um die ersten $i$ Zeichen der Quelle
in die ersten $j$ Zeichen des Ziels umzuwandeln.

- $i$ ist der **Zeilenindex** und die Länge des Quellpräfixes.
- $j$ ist der **Spaltenindex** und die Länge des Zielpräfixes.

Für unser Beispiel wandeln wir `cart` in `coast` um.
Die initialisierte Matrix sieht so aus:

| Quelle ↓ / Ziel → | ε | c | o | a | s | t |
| --- | ---: | ---: | ---: | ---: | ---: | ---: |
| ε | 0 | 1 | 2 | 3 | 4 | 5 |
| c | 1 | 0 | 0 | 0 | 0 | 0 |
| a | 2 | 0 | 0 | 0 | 0 | 0 |
| r | 3 | 0 | 0 | 0 | 0 | 0 |
| t | 4 | 0 | 0 | 0 | 0 | 0 |

Die Matrix hat `len(source) + 1` Zeilen und `len(target) + 1` Spalten.

- In **Zeile 0** ist die Quelle leer. Wir brauchen $j$ Einfügungen, um
  das Zielpräfix der Länge $j$ zu erhalten.
- In **Spalte 0** ist das Ziel leer. Wir brauchen $i$ Löschungen, um das
  Quellpräfix der Länge $i$ in die leere Zeichenkette umzuwandeln.

Damit können wir beide Ränder sofort ausfüllen. Die Nullen in den inneren
Zellen sind zunächst Platzhalter. Ihre Werte berechnen wir im nächsten Schritt.


```python
def initialize_matrix(source: str, target: str) -> list[list[int]]:

    D = []
    for _ in range(len(source) + 1):
        D.append([0] * (len(target) + 1))
    
    for i in range(len(source) + 1):
        D[i][0] = i

    for j in range(len(target) + 1):
        D[0][j] = j

    return D
```

```python
def print_matrix(matrix: list[list[int]], source: str, target: str) -> None:
    print('   ' + ' '.join(f'{character:>2}' for character in 'ε' + target))

    for character, row in zip('ε' + source, matrix):
        values = ' '.join(f'{value:>2}' for value in row)
        print(f'{character:>2} {values}')
```

```python
source = 'cart'
target = 'coast'

D = initialize_matrix(source, target)

print_matrix(D, source, target)
```

### 6.2 Eine Zelle berechnen

Um eine innere Zelle `D[i][j]` mit $i,j > 0$ zu berechnen, benötigen wir
drei bereits berechnete Nachbarzellen: **oben**, **links** und **diagonal
oben links**.

Jede Nachbarzelle enthält die Distanz für ein kleineres Teilproblem. 
Für die aktuelle Zelle `D[i][j]` berechnen wir drei Kandidaten:

1. **Oben:** `D[i-1][j] + 1`
2. **Links:** `D[i][j-1] + 1`
3. **Oben links:** `D[i-1][j-1] + c`

Für $c$ vergleichen wir das letzte Zeichen des Quellpräfixes `source[i-1]`
mit dem letzten Zeichen des Zielpräfixes `target[j-1]`:

- Sind beide Zeichen gleich, ist $c = 0$.
- Sind sie verschieden, ist $c = 1$.


In die aktuelle Zelle tragen wir den kleinsten der drei Werte ein.

$$
D_{i,j} = \min\bigl(D_{i-1,j} + 1,\; D_{i,j-1} + 1,\; D_{i-1,j-1} + c\bigr)
$$

**Beispiel `cart` → `coast`.**
In `D[1][1]` vergleichen wir `c` mit `c`:

- Oben: `1 + 1 = 2`
- Links: `1 + 1 = 2`
- Diagonal: `0 + 0 = 0`, weil die Zeichen gleich sind.

Das Minimum ist **0**. Wir tragen also `D[1][1] = 0` ein.

Bei `D[1][2]` vergleichen wir `c` mit `co`. Hier ist Einfügen
am günstigsten:

- Oben: `2 + 1 = 3`
- Links: `0 + 1 = 1`
- Diagonal: `1 + 1 = 2`, weil `c` und `o` verschieden sind

Das Minimum ist **1**. Wir tragen also `D[1][2] = 1` ein.


```python
def calculate_cell(D: list[list[int]], source: str, target: str, i: int, j: int) -> int:
    
    if i < 1 or j < 1:
        raise ValueError('i and j must be at least 1')

    # TODO: Calculate the minimum cost using the neighboring top, left, and top-left cells.
    top = D[i-1][j] + 1
    left = D[i][j-1] + 1
    diagonal = D[i-1][j-1] + (1 if source[i-1] != target[j-1] else 0)

    return min(top, left, diagonal)
    
    raise NotImplementedError('Implement calculate_cell')
```

### 6.3 Die ganze Matrix füllen

Wir wenden dieselbe Regel auf alle inneren Zellen an, Zeile für Zeile
von links nach rechts. So sind die drei Nachbarzellen schon berechnet,
wenn wir sie brauchen.

Das ist **dynamische Programmierung**: Wir lösen zuerst kleinere
Teilprobleme, speichern ihre Ergebnisse und verwenden sie für größere
Teilprobleme wieder.

Die Zelle ganz rechts unten enthält anschließend die Levenshtein-Distanz
zwischen den vollständigen Zeichenketten.

```python
def levenshtein_matrix(source: str, target: str) -> list[list[int]]:
    D = initialize_matrix(source, target)

    for i in range(1, len(source) + 1):
        for j in range(1, len(target) + 1):
            D[i][j] = calculate_cell(D, source, target, i, j)

    return D


def levenshtein_distance(source: str, target: str) -> int:
    D = levenshtein_matrix(source, target)
    return D[-1][-1]


D = levenshtein_matrix(source, target)

print_matrix(D, source, target)

print(f'\nLevenshtein distance: ', D[-1][-1])
```

### Testfälle

```python
import unittest
check = unittest.TestCase()

check.assertEqual(levenshtein_distance('', ''), 0)
check.assertEqual(levenshtein_distance('cat', ''), 3)
check.assertEqual(levenshtein_distance('', 'cat'), 3)
check.assertEqual(levenshtein_distance('cat', 'cat'), 0)

check.assertEqual(levenshtein_distance('cat', 'cats'), 1)
check.assertEqual(levenshtein_distance('cats', 'cat'), 1)
check.assertEqual(levenshtein_distance('cat', 'cut'), 1)
check.assertEqual(levenshtein_distance('cat', 'cuts'), 2)
check.assertEqual(levenshtein_distance('cart', 'coast'), 2)
check.assertEqual(levenshtein_distance('cat', 'dog'), 3)
check.assertEqual(levenshtein_distance('cat', 'actor'), 4)
check.assertEqual(levenshtein_distance('intention', 'execution'), 5)

print('Levenshtein tests passed.')
```
