# Wie lese ich eine Markdown-Datei Zeile für Zeile ein, um sie auf Auffälligkeiten zu prüfen?

Wählt den Ordner mit den zu prüfenden Markdown-Dateien. Für jede Datei rufen wir eine Prüffunktion pro Zeile auf. Das Ergebnis hat bereits die drei Felder der späteren Meldung: `id`, `line` und `type`.

```python
from pathlib import Path
from random import Random

folder = Path('../examples/anomaly_markdown')
randomizer = Random(42)
anomaly_types = ['spelling', 'format', 'extra_word', 'advertising']


def check_line(line: str) -> str:
    return randomizer.choice(anomaly_types)
```

`check_line` ist ein Platzhalter. Er nimmt den Zeilentext entgegen und gibt zum Ausprobieren einen zufälligen Typ zurück, auch bei einer Leerzeile. Der Startwert 42 macht einen vollständigen Durchlauf wiederholbar. Ersetzt den Platzhalter später durch eure Prüfung und speichert die tatsächlich erkannten Befunde.

```python
for path in sorted(folder.glob('*.md')):
    document_id = path.stem.removeprefix('epd_').removesuffix('_with_anomalies')
    with path.open(encoding='utf-8') as file:
        for line_number, line in enumerate(file, start=1):
            anomaly_type = check_line(line)
            print(f'id={document_id}  line={line_number}  type={anomaly_type}')
            if line_number == 5:
                print('Skip to next document\n')
                break
```

`glob('*.md')` findet die Markdown-Dateien direkt im gewählten Ordner. Aus `epd_101_with_anomalies.md` wird die ID `101`. Zeile 2 der Datei ist leer und erhält trotzdem ihre eigene Nummer. Die Vorschau endet für EPD 101 bei Zeile 5 und beginnt für EPD 102 wieder bei 1. Entfernt den gesamten Block ab `if line_number == 5:` einschließlich `print` und `break`, um jede Datei bis zum Ende zu prüfen.
