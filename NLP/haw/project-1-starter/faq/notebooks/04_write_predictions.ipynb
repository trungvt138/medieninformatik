# Wie schreibe ich meine Meldungen als CSV-Datei?

Eine Meldung besteht aus EPD-ID, Zeilennummer und Fehlertyp. Ersetzt die beiden Beispielmeldungen durch eure eigenen. Die Schreibzelle erzeugt `demo_predictions.csv` neben dem Notebook und ersetzt sie bei jedem Lauf. Verwendet für eure tatsächliche Validierung einen eigenen Dateinamen.

```python
import csv
from pathlib import Path

output_path = Path('demo_predictions.csv')
predictions = [
    {'id': '101', 'line': 3, 'type': 'spelling'},
    {'id': '102', 'line': 3, 'type': 'format'},
]
```

```python
with output_path.open('w', encoding='utf-8', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=['id', 'line', 'type'])
    writer.writeheader()
    writer.writerows(predictions)

print(output_path.read_text(encoding='utf-8'))
```

Setzt für einen zweiten Versuch `predictions = []`. Die Kopfzeile bleibt auch bei null Meldungen erhalten. Für einen Probelauf mit der [Evaluation](05_evaluate.ipynb) setzt ihr dort `predictions = Path('demo_predictions.csv')` und verwendet die mitgelieferten Beispiel-Sollbefunde.
