# Wie verwende ich Code aus einer eigenen Python-Datei im Notebook?

Eine Python-Datei mit Funktionen heißt beim Import **Modul**. In diesem Ordner liegt [evaluation.py](evaluation.py). Ihr könnt die Datei öffnen und die Funktion `ratio` darin nachlesen. Sie berechnet einen Anteil und verwendet für einen Nenner von null die vereinbarte Ausgabe 0.

So geht ihr bei eigenem Code vor:

1. Legt die Funktion in einer Datei mit der Endung `.py` neben eurem Notebook ab und speichert sie.
2. Importiert sie mit `from dateiname import funktionsname`. Im Import steht der Dateiname ohne `.py`.
3. Ruft die Funktion wie eine im Notebook definierte Funktion auf. Für `ratio` sieht das so aus:

```python
from evaluation import ratio

hits = 3
predictions = 4
print(ratio(hits, predictions))
```

Drei Treffer unter vier Meldungen ergeben eine Precision von 0,75. Probiert auch den Fall ohne Meldungen.

```python
print(ratio(0, 0))
```

Wenn ihr eine importierte Funktion ändert, speichert die Python-Datei, startet den Kernel neu und führt die Zellen erneut aus. Das [Notebook zur Evaluation](05_evaluate.ipynb) importiert auf dieselbe Weise die vollständige Funktion `evaluate` aus derselben Datei.
