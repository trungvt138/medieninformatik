# Wie lese ich den vollständigen Text jeder EPD?

Wählt einen Ordner mit Markdown-Dateien. Für eine Wortliste aus Trainingsoriginalen eignet sich `training/markdown`. Ihr erhaltet ein Wörterbuch, in dem zu jedem Dateinamen der gesamte Text steht.

```python
from pathlib import Path

folder = Path('../examples/anomaly_markdown')
paths = sorted(folder.glob('*.md'))

documents = {}
for path in paths:
    text = path.read_text(encoding='utf-8')
    documents[path.name] = text
    print(f'{path.name}: {len(text)} characters')
```

Gebt nur beim ersten Dokument einen Auszug aus. Ändert die Länge der Vorschau, wenn ihr mehr vom Text sehen möchtet.

```python
if paths:
    first_name = paths[0].name
    print(f'Preview of {first_name}:')
    print(documents[first_name][:800])
else:
    print('No Markdown files found.')
```

Die Vorschau zeigt höchstens 800 Zeichen. Das Wörterbuch enthält weiterhin jeden vollständigen Text. In der [Wörterbuchübung aus der Vorlesung](https://git.haw-hamburg.de/nlp-wise-26/students/02-1_word-similarity/-/blob/main/02_dictionary_check.ipynb) könnt ihr nachsehen, wie daraus eine Wortliste entsteht.
