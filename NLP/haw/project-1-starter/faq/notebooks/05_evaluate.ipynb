# Welche Kennzahlen erreicht mein Verfahren?

Setzt zwei Pfade: `predictions` auf eure erzeugte Ergebnis-CSV und `reference` auf die zugehörige `findings.csv`. Für die Projektvalidierung stammt diese aus dem Ordner `validation`. Die beiden Beispieldateien sind mitgeliefert. Wie der Import der [Auswertungsfunktion](evaluation.py) funktioniert, zeigt die [FAQ zu eigenen Python-Dateien](06_import_python.ipynb).

```python
from pathlib import Path
from evaluation import evaluate

predictions = Path('../examples/predictions.csv')
reference = Path('../examples/findings.csv')
```

```python
rows = evaluate(predictions, reference)
header = '{:<12} {:>4} {:>4} {:>4} {:>10} {:>8} {:>8}'
print(header.format('type', 'TP', 'FP', 'FN', 'precision', 'recall', 'F1'))
for row in rows:
    print('{type:<12} {TP:>4} {FP:>4} {FN:>4} {precision:>10.3f} {recall:>8.3f} {F1:>8.3f}'.format_map(row))
```

TP sind Treffer, FP Fehlalarme, FN übersehene Befunde. Precision ist der Trefferanteil unter den Meldungen, Recall der gefundene Anteil der Sollbefunde. F1 ist ihr harmonisches Mittel. Berichtet die Werte je Fehlertyp. Die zusätzliche Zeile `overall` berechnet die Kennzahlen aus den über alle Typen summierten TP, FP und FN (Mikro-Mittelung). Bei einem Nenner von null gibt die Funktion 0 aus.

Im Beispiel zählt die doppelte Meldung zu EPD 101, Zeile 3 einmal.
