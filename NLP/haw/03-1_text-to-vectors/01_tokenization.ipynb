## 1. Tokenisierungen vergleichen

### 1.1 Mit `split()` an Leerzeichen trennen

Verwende [`str.split()`](https://docs.python.org/3.3/library/stdtypes.html#str.split), um die Review in eine Liste von Wörtern zu zerlegen.

```python
def tokenize1(text: str) -> list[str]:
    return text.split()
    
    raise NotImplementedError()


split_tokens = tokenize1(review)

print('split():', split_tokens)
```

### 1.2 Wörter mit Regex finden

Verwende `re.findall()` und ein Pattern für zusammenhängende englische Buchstaben (`A–Z` und `a–z`). Behalte die Groß- und Kleinschreibung bei.

```python
def tokenize2(text: str) -> list[str]:
    return re.findall(r'\w+', text)
    raise NotImplementedError()


regex_tokens = tokenize2(review)

print('Regex:  ', regex_tokens)
```

### 1.3 Mehrere Tokenregeln kombinieren

Welche Tokens erwartest du für `Dr.`, `5/7` und `stars!`? Führe dann die Zelle aus. Python probiert die Alternativen (`|`) an derselben Textposition der Reihe nach. Deshalb stehen die Regeln für Abkürzungen, Brüche und Zahlen vor der allgemeinen Wortregel.

Das Beispiel ist eine angepasste Fassung aus dem [NLTK-Buch, Abschnitt 3.7](https://www.nltk.org/book/ch03.html#regular-expressions-for-tokenizing-text) von Bird, Klein und Loper (2009).

```python
pattern = r'''(?x)                        # Ignore formatting spaces and allow comments.
    (?:[A-Z]\.)+                          # One or more capital letters, each followed by a dot.
  | (?:Dr|Mr|Mrs|Ms|Prof)\.               # Listed titles with their trailing dot.
  | \d+/\d+                               # One or more digits on each side of a slash.
  | \$?\d+(?:\.\d+)?%?                    # Digits with an optional dollar sign, decimals and percent sign.
  | \w+(?:-\w+)*                          # Letters, digits or underscore, with optional hyphenated parts.
  | \.\.\.                                # Three literal dots as one token.
  | [][.,;"'!?():_`–—-]                   # One listed punctuation mark. The final hyphen is literal.
'''

def tokenize3(text: str, pattern: str) -> list[str]:
    return re.findall(pattern, text)


nltk_tokens = tokenize3(review, pattern)

print('NLTK Regex:', nltk_tokens)
```
