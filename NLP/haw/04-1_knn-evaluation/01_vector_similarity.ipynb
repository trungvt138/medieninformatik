## 4. Eigene Texte vergleichen

Die fertige Funktion bekommt eine Suchanfrage (Query) und eine Liste von Texten. Sie lernt die Wortspalten aus der Liste und gibt den Text mit der kleinsten Kosinusdistanz zurück. Bei Gleichstand gewinnt der erste. Die Query muss mindestens ein Wort aus dem Vokabular enthalten.

```python
def most_similar_text(query: str, texts: list[str]) -> str:
    if not texts:
        raise ValueError('Provide at least one candidate text.')
    vectorizer = CountVectorizer()
    candidates = vectorizer.fit_transform(texts)
    query_vector = vectorizer.transform([query])
    if query_vector.nnz == 0:
        raise ValueError('Use at least one word from the candidate vocabulary.')
    distances = cosine_distances(query_vector, candidates)[0]
    return texts[int(distances.argmin())]
```

Welcher Text gewinnt für D? Probiere anschließend `Flat direction.` als Query. Ergänze einen eigenen Kandidaten und prüfe, ob sich der Gewinner ändert.

```python
candidates = [A, B]
query = D
print(most_similar_text(query, candidates))
```
