## 1. Filmreviews vorbereiten

A, B und C liefern das Vokabular. D ist eine neue Review. Welches Wort aus D fehlt in A bis C?

```python
from sklearn.feature_extraction.text import CountVectorizer

A = 'Great acting. Great direction. Great soundtrack.'
B = 'Flat acting.'
C = 'Great acting.'
known_texts = [A, B, C]

D = 'Great acting overall.'
new_texts = [D]
```

## 2. Wortspalten aus A, B und C lernen

Mit `vectorizer.fit(known_texts)` lernt der Vektorisierer aus A, B und C, welches Wort zu welcher Spalte gehört. Der Aufruf gibt den angepassten Vektorisierer zurück. [Dokumentation zu fit()](https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html#sklearn.feature_extraction.text.CountVectorizer.fit)

Hier wollen wir die Wörter in A, B und C auch gleich zählen. `fit_transform(known_texts)` lernt das Vokabular und gibt die Zählmatrix für dieselben Texte zurück. Jede Zeile steht für eine Review, jede Spalte für ein Wort. [Dokumentation zu fit_transform()](https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html#sklearn.feature_extraction.text.CountVectorizer.fit_transform)

```python
vectorizer = CountVectorizer()
known_vectors = vectorizer.fit_transform(known_texts)
print(vectorizer.get_feature_names_out())
```

<a id="sparse"></a>

### 2.1 Dünn besetzte Darstellung (sparse)

Eine Review enthält meist nur wenige Wörter aus dem gesamten Vokabular. Der `CountVectorizer` speichert seine Zählmatrix daher im CSR-Format (*Compressed Sparse Row*). Es speichert die von null verschiedenen Zählwerte mit Angaben zu ihrer Zeile und Spalte. Die Ausgabe zeigt dir für jeden gespeicherten Wert die Koordinaten `(Zeile, Spalte)`. Beide Indizes beginnen bei 0. [scikit-learn zu dünn besetzten Textmatrizen](https://scikit-learn.org/stable/modules/feature_extraction.html#sparsity)

```python
print('Matrix type:', type(known_vectors).__name__)
print(known_vectors)
```

### 2.2 Dichte Darstellung (dense)

Suche in der vorigen Ausgabe den Eintrag `(0, 3)  3`. Zeile 0 gehört zu Review A, Spalte 3 zum Wort *great*. Es kommt dreimal vor. Mit `toarray()` siehst du dieselben Zählwerte als vollständige Tabelle, einschließlich aller Nullen. Wo findest du die 3 wieder?

```python
print(known_vectors.toarray())
```

## 3. Die neue Review D vektorisieren

Mit `transform(new_texts)` zählt der Vektorisierer die Wörter der neuen Review in den bereits gelernten Spalten. Wörter außerhalb des Vokabulars werden übersprungen. [Dokumentation zu transform()](https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html#sklearn.feature_extraction.text.CountVectorizer.transform)

Welche Zahlen erwartest du für D? Vergleiche deine Vorhersage mit der nächsten Ausgabe.

```python
new_vector = vectorizer.transform(new_texts)
print('C:', known_vectors.toarray()[2])
print('D:', new_vector.toarray()[0])
```

Warum verändert *overall* hier keinen Zählwert? Was würde schiefgehen, wenn du für D ein eigenes Vokabular lernen würdest?
