```python
from cpmpy import *

# F = ['rot', 'weiß', 'grün', 'gelb', 'blau']
# N = ['Brite', 'Schwede', 'Däne', 'Norweger', 'Deutsche']
# G = ['Tee', 'Milch', 'Bier', 'Wasser', 'Kaffee']
# T = ['Vogel', 'Katze', 'Hund', 'Pferde', 'Maus']
# Z = ['Pall-Mall', 'Prince', 'Blends', 'BlueMaster', 'Dunhill']
# H = [a for a in range(5)]

rot, weiss, gruen, gelb, blau = intvar(0, 4, shape=5)
brite, schwede, daene, norweger, deutsche = intvar(0, 4, shape=5)
tee, milch, bier, wasser, kaffee = intvar(0, 4, shape=5)
vogel, katze, hund, pferde, fisch = intvar(0, 4, shape=5)
pallmall, prince, blends, bluemaster, dunhill = intvar(0, 4, shape=5)

model = Model()

model += AllDifferent(rot, weiss, gruen, gelb, blau)
model += AllDifferent(brite, schwede, daene, norweger, deutsche)
model += AllDifferent(tee, milch, bier, wasser, kaffee)
model += AllDifferent(vogel, katze, hund, pferde, fisch)
model += AllDifferent(pallmall, prince, blends, bluemaster, dunhill)

model += (brite == rot)
model += (schwede == hund)
model += (daene == tee)
model += (gruen == weiss - 1)
model += (gruen == kaffee)
model += (pallmall == vogel)
model += (gelb == dunhill)
model += (milch == 2)
model += (norweger == 0)
model += (abs(blends - katze) == 1)
model += (abs(pferde - dunhill) == 1)
model += (bluemaster == bier)
model += (deutsche == prince)
model += (blau == 1)
model += (abs(blends - wasser) == 1)

labels = {
    rot:'rot', weiss:'weiss', gruen:'gruen', gelb:'gelb', blau:'blau',
    brite:'Brite', schwede:'Schwede', daene:'Däne', norweger:'Norweger', deutsche:'Deutsche',
    tee:'Tee', milch:'Milch', bier:'Bier', wasser:'Wasser', kaffee:'Kaffee',
    vogel:'Vogel', katze:'Katze', hund:'Hund', pferde:'Pferde', fisch:'Fisch',
    pallmall:'Pall-Mall', prince:'Prince', blends:'Blends', bluemaster:'BlueMaster', dunhill:'Dunhill',
}

def pretty_print():
    houses = [[] for _ in range(5)]
    for var, name in labels.items():
        houses[var.value()].append(name)
    for i, attrs in enumerate(houses):
        print(f"Haus {i+1}: {', '.join(attrs)}")
    print()

model.solveAll(display=pretty_print)

print(rot.value(), weiss.value(), gruen.value(), gelb.value(), blau.value())
print(brite.value(), schwede.value(), daene.value(), norweger.value(), deutsche.value())
print(tee.value(), milch.value(), bier.value(), wasser.value(), kaffee.value())
print(vogel.value(), katze.value(), hund.value(), pferde.value(), fisch.value())
print(pallmall.value(), prince.value(), blends.value(), bluemaster.value(), dunhill.value())
```
