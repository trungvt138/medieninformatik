### Aufgabe 7 – Gerade Zahlen zählen

Schreibe eine Funktion `count_even(numbers)`, die eine Liste von ganzen Zahlen (`numbers`) bekommt und zurückgibt, wie viele davon **gerade** sind.

Beispiel:
- `count_even([1, 2, 3, 4])` → `2`

👉 Ergänze den Code.


````python
def count_even(numbers):
    """Gibt die Anzahl gerader Zahlen in der Liste numbers zurück."""
    # TODO: hier deine Lösung
    count = 0
    for n in numbers:
        if n % 2 == 0:
            count += 1
    return count

# Tests:
assert count_even([1, 2, 3, 4]) == 2
assert count_even([]) == 0
assert count_even([2, 4, 6]) == 3
assert count_even([1, 3, 5]) == 0
print("Aufgabe 7: Tests bestanden ✅")
````

### Aufgabe 9 – Einfache Klasse

Schreibe eine einfache Klasse `Person` mit:

- einem Konstruktor `__init__(self, name, age)`, der die Attribute `name` und `age` speichert,
- einer Methode `greet(self)`, die einen String zurückgibt, z. B. `"Hallo, ich heiße Alice."`.

👉 Ergänze den Code.


````python
class Person:
    # TODO: Konstruktor und greet-Methode implementieren
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hallo, ich heiße {self.name}."

# Tests:
p = Person("Alice", 21)
assert p.name == "Alice"
assert p.age == 21
assert "Alice" in p.greet()
print("Aufgabe 9: Tests bestanden ✅")
````
