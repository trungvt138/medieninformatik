## 2. Variablen & Datentypen

**Java:**
```java
int x = 5;
double y = 3.14;
String name = "Alex";
```

Du musst den Datentyp vorher angeben.

**Python:**
```python
x = 5

```

- Keine Typdeklaration nötig, Python erkennt den Typ automatisch.

````python
# Beispiel: Variablen in Python

x = 5
y = 3.14
name = "Alex"

print("x:", x)
print("y:", y)
print("name:", name)
print("Typ von x:", type(x))
print("Typ von y:", type(y))
print("Typ von name:", type(name))
````

## 3. If-Abfragen

**Java:**
```java
if (x > 10) {
    System.out.println("x ist größer als 10");
} else {
    System.out.println("x ist nicht größer als 10");
}
```

**Python:**
```python
if x > 10:
    print("x ist größer als 10")
else:
    print("x ist nicht größer als 10")
```

Unterschiede:
- Keine `{}`-Klammern – der Block wird durch **Einrückung** (Indentation) definiert.
- Doppelpunkt `:` am Ende der `if`- und `else`-Zeile.
- Kein `;` am Zeilenende.

````python
x = 7

if x > 10:
    print("x ist größer als 10")
else:
    print("x ist nicht größer als 10")
````

## 4. Schleifen

### for-Schleife

**Java:**
```java
for (int i = 0; i < 5; i++) {
    System.out.println(i);
}
```

**Python:**
```python
for i in range(5):
    print(i)
```

- `range(5)` erzeugt die Zahlen `0, 1, 2, 3, 4`.
- Kein Inkrement `i++` nötig, das übernimmt `range`.

````python
for i in range(5):
    print("i =", i)
````

### while-Schleife

**Java:**
```java
int i = 0;
while (i < 3) {
    System.out.println(i);
    i++;
}
```

**Python:**
```python
i = 0
while i < 3:
    print(i)
    i += 1
```

````python
i = 0
while i < 3:
    print("i =", i)
    i += 1
````

## 5. Listen vs. Arrays / ArrayList

**Java (Array):**
```java
int[] zahlen = {1, 2, 3};
```

**Java (ArrayList):**
```java
ArrayList<Integer> liste = new ArrayList<>();
liste.add(1);
liste.add(2);
```

**Python (Liste):**
```python
zahlen = [1, 2, 3]
zahlen.append(4)
```

- Die Standard-Datenstruktur für geordnete Sammlungen in Python ist die **Liste** (`list`).
- Listen können verschiedene Typen gleichzeitig enthalten (z. B. Zahlen und Strings).

````python
zahlen = [1, 2, 3]
zahlen.append(4)
print(zahlen)
print("Erstes Element:", zahlen[0])
print("Länge der Liste:", len(zahlen))
````

## 6. Funktionen vs. Methoden

**Java (Methode in einer Klasse):**
```java
public static int add(int a, int b) {
    return a + b;
}
```

**Python (Funktion):**
```python
def add(a, b):
    return a + b
```

- Kein `public`, `static` oder Typangabe nötig.
- Funktionen können überall (auch außerhalb von Klassen) definiert werden.

````python
def add(a, b):
    return a + b

print("2 + 3 =", add(2, 3))
````

## 7. Klassen – Java-Klasse vs. Python-Klasse

**Java:**
```java
public class Person {
    private String name;
    private int age;

    public Person(String name, int age) {
        this.name = name;
        this.age = age;
    }

    public String greet() {
        return "Hallo, ich heiße " + name;
    }
}
```

**Python:**
```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hallo, ich heiße {self.name}"
```

- Der Konstruktor heißt in Python immer `__init__`.
- `self` entspricht grob `this` in Java.
- Attribute werden einfach über `self.name` zugewiesen.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hallo, ich heiße {self.name} und bin {self.age} Jahre alt."

p = Person("Alex", 21)
print(p.greet())
````

Schreibe eine kleine Klasse `Song` mit:
- Attributen: `title` (String), `length_seconds` (int)
- einer Methode `describe()`, die z. B. zurückgibt:
  
  `"Song: <Titel>, Dauer: <Min:Sek>"`

Hinweis: Du kannst Minuten und Sekunden etwa so berechnen:

```python

seconds = length_seconds % 60
```

````python
# TODO: Klasse Song implementieren
class Song:
    def __init__(self, title, length_seconds):
        self.title = title
        self.length_seconds = length_seconds

    def describe(self):
        return f"Song: {self.title}, Dauer: {self.length_seconds // 60}:{self.length_seconds % 60}"

s = Song("Smooth Operator", 259)
print(s.describe())
````
