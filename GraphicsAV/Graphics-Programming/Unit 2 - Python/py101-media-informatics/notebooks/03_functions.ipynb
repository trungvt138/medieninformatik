## 1. Warum Funktionen?

Ohne Funktionen landet schnell alles in einem großen Block – schwer zu lesen, schwer zu testen.

Funktionen helfen dir dabei:

- Code wiederzuverwenden
- komplexe Probleme in kleinere Teile zu zerlegen
- den Code lesbarer und testbarer zu machen

Ein typisches Muster:

```python
def mach_etwas(eingabe):
    # 1. Eingabe verarbeiten
    # 2. Ergebnis berechnen
    return ergebnis
```

## 4. Default-Parameter & benannte Argumente

Du kannst Funktionen Parameter mit Standardwerten geben.

```python
def greet_user(name, language="de"):
    if language == "de":
        print("Hallo", name)
    elif language == "en":
        print("Hello", name)
```

Aufruf:

```python
greet_user("Alex")          # nutzt language="de"
greet_user("Alex", "en")    # überschreibt language
greet_user(name="Alex")     # benanntes Argument
```

````python
def greet_user(name, language="de"):
    if language == "de":
        print("Hallo", name)
    elif language == "en":
        print("Hello", name)
    else:
        print("Hi", name)

greet_user("Alex")
greet_user("Jordan", "en")
greet_user(name="Taylor", language="fr")
````

### Mini-Übung 3 – Lautstärke-Anzeige

Schreibe eine Funktion `print_volume(level, max_level=10)`, die eine einfache Lautstärke-Anzeige ausgibt, z. B.:

```text
[#####-----] 5/10
```

Hinweise:

- Anzahl `#` entspricht `level`
- Anzahl `-` entspricht `max_level - level`
- Nutze z. B. String-Multiplikation: `"#" * 5` ergibt `"#####"`.

````python
# TODO: Mini-Übung 3 – print_volume
def print_volume(level, max_level=10):
    return f"[{'#'*level}{'-'*(max_level - level)}] {level}/{max_level}"

print(print_volume(5))
````

## 5. Scope: lokale vs. globale Variablen

Variablen, die **in einer Funktion** definiert werden, sind **lokal**. Von außen kannst du sie nicht direkt verwenden.

```python
def foo():
    x = 42  # lokal
    print("In foo:", x)

foo()
print(x)  # Fehler!
```

Variablen außerhalb von Funktionen nennt man **global**. Sie sind im gesamten Modul sichtbar (mit Vorsicht zu verwenden).

````python
value = 10  # globale Variable

def multiply_by_two(n):
    result = n * 2  # lokal
    return result

print("value:", value)
print("value * 2:", multiply_by_two(value))

# Die folgende Zeile würde einen Fehler auslösen, weil result lokal ist:
# print(result)
````

## 6. Funktionen kombinieren

Statt eine riesige Funktion zu schreiben, ist es oft besser, **mehrere kleine Funktionen** zu kombinieren.

Beispiel: Wir wollen ein kleines Text-UI für Songinfos.

- Funktion zum Einlesen der Eingaben
- Funktion zur Umrechnung der Dauer
- Funktion, die den finalen Text baut

Das könnte so aussehen:

````python
def read_song_info():
    title = input("Songtitel: ")
    seconds_text = input("Dauer in Sekunden: ")
    seconds = int(seconds_text)
    return title, seconds

def format_duration(seconds):
    minutes = seconds // 60
    rest = seconds % 60
    return f"{minutes}:{rest:02d} Minuten"

def describe_song(title, seconds):
    duration_str = format_duration(seconds)
    return f"Song: {title}, Dauer: {duration_str}"

# Zusammenspiel:
# (auskommentiert, damit das Notebook nicht ständig input verlangt)

# t, s = read_song_info()
# print(describe_song(t, s))
````

## 7. Abschlussaufgabe – Medien-Score berechnen

Schreibe ein kleines System, das einen **"Media Engagement Score"** für eine Person berechnet.

Du bekommst (z. B. per Eingabe oder als feste Variablen):

- `stunden_musik_pro_woche` (float)
- `stunden_games_pro_woche` (float)
- `stunden_video_pro_woche` (float)

Baue folgende Funktionen:

1. `total_hours(music, games, video)` → gibt die Summe der Stunden zurück.
2. `media_score(music, games, video)` → gibt einen Score zwischen 0 und 100 zurück.
   - z. B.: `score = min(100, total_hours * 5)`
3. `classify_score(score)` → gibt einen Text zurück, z. B.:
   - unter 10: `"Kaum aktiv"`
   - 10–30: `"Moderat aktiv"`
   - 30–60: `"Sehr aktiv"`
   - über 60: `"Extrem aktiv"`

Schreibe ein kleines Hauptprogramm, das feste Beispielwerte nutzt, z. B.:

```python
music = 5.0
games = 3.0
video = 7.5
```

und dann den Score sowie die Klassifikation ausgibt.

````python
# TODO: Abschlussaufgabe – Media Engagement Score
def total_hours(music, games, video):
    return music + games + video

def media_score(music, games, video):
    return min(100, total_hours(music, games, video) * 5)

def classify_score(score):
    if score < 10:
        print("Kaum aktiv")
    elif score >= 10 and score < 30:
        print("Moderat aktiv")
    elif score >= 30 and score < 60:
        print("Sehr aktiv")
    else:
        print("Extrem aktiv")

classify_score(media_score(5.0, 3.0, 7.5))
````
