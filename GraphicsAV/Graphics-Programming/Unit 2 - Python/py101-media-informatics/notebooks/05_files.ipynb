## 1. Datei öffnen: Grundprinzip

Das Grundmuster zum Lesen/Schreiben einer Datei sieht so aus:

```python
f = open("data/beispiel.txt", "r", encoding="utf-8")
inhalt = f.read()
f.close()
```

- Erster Parameter: Pfad zur Datei
- Zweiter Parameter: Modus (`"r"` = lesen, `"w"` = schreiben, `"a"` = anhängen)
- `encoding="utf-8"` ist sinnvoll für Umlaute und Sonderzeichen
- Wichtig: `close()` nicht vergessen – oder besser direkt `with` verwenden (siehe unten).

## 2. Textdateien lesen

Du kannst ganze Dateien auf einmal oder Zeile für Zeile lesen.    
(Das `try`....`except`-Konstrukt wird in Notebook 06 ausführlicher erklärt.) 

````python
# Beispiel: Datei komplett lesen (wenn vorhanden)
# Diese Zelle funktioniert nur, wenn du vorher eine passende Datei angelegt hast.

try:
    with open("data/example.txt", "r", encoding="utf-8") as f:
        content = f.read()
    print("Dateiinhalt:")
    print(content)
except FileNotFoundError:
    print("Die Datei data/example.txt existiert noch nicht. Lege sie im Projektordner an.")
````

Du kannst auch Zeile für Zeile lesen – praktisch für Logs oder Listen.

````python
# Zeilenweise lesen (falls Datei existiert)
try:
    with open("data/example.txt", "r", encoding="utf-8") as f:
        print("Zeilenweise Ausgabe:")
        for line in f:
            print("-", line.strip())
except FileNotFoundError:
    print("Noch keine example.txt vorhanden.")
````

## 3. Textdateien schreiben

Mit Modus `"w"` überschreibst du eine Datei (oder legst sie neu an), mit `"a"` hängst du Text an.

```python
with open("data/out.txt", "w", encoding="utf-8") as f:
    f.write("Hallo Datei!\n")
```

````python
# Beispiel: Datei neu schreiben
from datetime import datetime

with open("data/log.txt", "w", encoding="utf-8") as f:
    f.write("Logdatei gestartet\n")
    f.write(f"Erstellt am: {datetime.now()}\n")

print("log.txt wurde neu geschrieben.")
````

````python
# Beispiel: an eine Datei anhängen
with open("data/log.txt", "a", encoding="utf-8") as f:
    f.write("Weitere Zeile im Log.\n")

print("Weitere Zeile an log.txt angehängt.")
````

## 4. Sicher arbeiten mit `with`

`with open(...) as f:` ist die **empfohlene** Art, mit Dateien zu arbeiten.

Warum?

- Datei wird automatisch geschlossen, auch bei Fehlern
- weniger Tippaufwand
- sauberer Code

Merke dir dieses Muster:

```python
with open(pfad, modus, encoding="utf-8") as f:
    # hier mit f arbeiten
```

## 5. CSV-Dateien – einfache Tabellendaten

CSV bedeutet **Comma-Separated Values** – also Komma-getrennte Werte.

Beispiel für eine `playlists.csv`:

```text
title,artist,length_seconds
Song A,Artist 1,197
Song B,Artist 2,220
```

Python hat dafür das eingebaute `csv`-Modul.

````python
import csv

# Beispiel: CSV schreiben
songs = [
    {"title": "Song A", "artist": "Artist 1", "length_seconds": 197},
    {"title": "Song B", "artist": "Artist 2", "length_seconds": 220},
]

fieldnames = ["title", "artist", "length_seconds"]

with open("data/playlist.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    for song in songs:
        writer.writerow(song)

print("playlist.csv wurde geschrieben.")
````

````python
# Beispiel: CSV lesen
import csv

try:
    with open("data/playlist.csv", "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        print("Songs aus playlist.csv:")
        for row in reader:
            title = row["title"]
            artist = row["artist"]
            length = int(row["length_seconds"])
            print(f"- {title} von {artist} ({length} Sekunden)")
except FileNotFoundError:
    print("playlist.csv wurde noch nicht erstellt.")
````
