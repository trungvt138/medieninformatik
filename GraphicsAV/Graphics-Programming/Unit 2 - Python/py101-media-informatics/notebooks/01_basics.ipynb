## 3. Datentypen: Zahlen, Strings, Booleans

Die wichtigsten einfachen Datentypen sind:

- **Integer** (`int`) – ganze Zahlen, z. B. `3`, `42`, `-10`
- **Float** (`float`) – Kommazahlen, z. B. `3.14`, `0.5`
- **String** (`str`) – Zeichenketten, z. B. `"Hallo"`
- **Boolean** (`bool`) – Wahrheitswerte: `True` oder `False`

Du kannst dir den Typ mit `type(...)` anzeigen lassen:

````python
x = 3
y = 3.14
text = "Hallo"
flag = True

print(x, "hat Typ", type(x))
print(y, "hat Typ", type(y))
print(text, "hat Typ", type(text))
print(flag, "hat Typ", type(flag))
````

## 4. Operatoren

### 4.1 Rechenoperatoren

- `+` Addition
- `-` Subtraktion
- `*` Multiplikation
- `/` Division (Ergebnis ist immer ein Float)
- `//` Ganzzahldivision
- `%` Rest (Modulo)
- `**` Potenz

````python
a = 7
b = 3

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
````

### Mini-Übung 4 – Sekunden in Minuten umrechnen

Schreibe ein kleines Stück Code, das eine Variable `sekunden` nimmt und ausgibt, wie viele Minuten und Sekunden das sind.

Beispiel: `sekunden = 197` → `3 Minuten und 17 Sekunden`.

````python
# TODO: Sekunden in Minuten und Sekunden umrechnen
def sekundenUmrechnen(sekunden):
    return (sekunden // 60, sekunden % 60)

sekundenUmrechnen(197)
````

### 4.2 Vergleichsoperatoren

- `==` gleich
- `!=` ungleich
- `<`, `>`, `<=`, `>=`

Sie liefern immer ein `True` oder `False` zurück.

````python
x = 5
y = 8

print("x == y:", x == y)
print("x != y:", x != y)
print("x < y:", x < y)
print("x >= 5:", x >= 5)
````

### 4.3 Logische Operatoren

- `and`
- `or`
- `not`

Beispiele:

````python
ist_student = True
hat_zeit = False

print("Student und Zeit:", ist_student and hat_zeit)
print("Student oder Zeit:", ist_student or hat_zeit)
print("Nicht Student:", not ist_student)
````

## 5. Typumwandlungen

Da `input(...)` immer einen String zurückgibt, musst du oft in `int` oder `float` umwandeln:

```python
zahl_text = input("Gib eine Zahl ein: ")
zahl = int(zahl_text)
print(zahl + 1)
```

Du kannst auch in String umwandeln:

```python
str(42)  # ergibt "42"
```

````python
zahl_text = input("Gib eine ganze Zahl ein: ")
zahl = int(zahl_text)
print("Du hast", zahl, "eingegeben. In fünf Jahren bist du", zahl + 5, "Jahre alt (wenn das dein Alter war).")
````
