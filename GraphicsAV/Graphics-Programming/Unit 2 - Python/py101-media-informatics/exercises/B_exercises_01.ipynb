## B1 – Punktestand normalisieren

Schreibe eine Funktion `normalize_score(points, max_points)`, die einen Prozentwert zwischen 0 und 100 zurückgibt.

Beispiel:

- `normalize_score(45, 60)` → `75.0`

Nutze die Funktion dann, um eine Liste von Punkteständen zu normalisieren.

````python
def normalize_score(points, max_points):
    if max_points <= 0:
        raise ValueError("max_points must be > 0")
    return points / max_points * 100

scores = [45, 30, 60]
max_points = 60

normalized = []
for s in scores:
    normalized.append(normalize_score(s, max_points))

print(normalized)

# Lösungshinweis:
# Prozent = points / max_points * 100
````

## B2 – Score klassifizieren

Schreibe eine Funktion `classify_score(score)`, die eine textuelle Kategorie zurückgibt:

- unter 25: "sehr wenig"
- 25–50: "wenig"
- 50–75: "mittel"
- 75–100: "viel"

````python
def classify_score(score):
    if score < 25:
        return "sehr wenig"
    elif score < 50:
        return "wenig"
    elif score < 75:
        return "mittel"
    else:
        return "viel"

for s in [10, 30, 55, 80]:
    print(s, "→", classify_score(s))

# Lösungshinweis:
# if / elif / else-Kette bauen.
````

## B3 – Medien-Sessions pro Nutzer aufaddieren

Gegeben ist eine Liste von Sessions:

```python
sessions = [
    {"user": "Alex", "minutes": 30},
    {"user": "Sam", "minutes": 20},
    {"user": "Alex", "minutes": 40},
]
```

Schreibe eine Funktion `total_minutes_per_user(sessions)`, die ein Dictionary zurückgibt:

```python
{"Alex": 70, "Sam": 20}
```

````python
sessions = [
    {"user": "Alex", "minutes": 30},
    {"user": "Sam", "minutes": 20},
    {"user": "Alex", "minutes": 40},
]

def total_minutes_per_user(sessions):
    totals = {}
    for s in sessions:
        name = s["user"]
        mins = s["minutes"]
        totals[name] = totals.get(name, 0) + mins
    return totals

print(total_minutes_per_user(sessions))

# Lösungshinweis:
# totals[name] = totals.get(name, 0) + mins
````

## B5 – Zeitformatierung für Playlist

Schreibe eine Funktion `format_duration(seconds)`, die einen String wie
`"3:17"` für 197 Sekunden zurückgibt.

Nutze sie, um eine Liste von Songs mit Titel und Dauer schön auszugeben.

````python
def format_duration(seconds):
    minutes = seconds // 60
    rest = seconds % 60
    return f"{minutes}:{rest:02d}"

songs = [
    {"title": "Song A", "length": 197},
    {"title": "Song B", "length": 245},
]

for song in songs:
    print(f"{song['title']} – {format_duration(song['length'])}")

# Lösungshinweis:
# Sekunden in min + sec aufspalten und führende Null formatieren.
````

## B6 – Sessions aus CSV lesen (falls vorhanden)

Versuche eine Datei `data/sessions.csv` im Format

```text
user,medium,minutes
Alex,Musik,30
Sam,Film,120
```

zu lesen und die Gesamtminuten pro Nutzer auszugeben.

Wenn die Datei nicht existiert, soll eine freundliche Meldung erscheinen.

````python
import csv

    path = "data/sessions.csv"

try:
    totals = {}
    with open(path, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            name = row["user"]
            minutes = int(row["minutes"])
            totals[name] = totals.get(name, 0) + minutes
    print("Gesamtminuten pro Nutzer aus CSV:")
    for name, mins in totals.items():
        print(f"- {name}: {mins} Minuten")
except FileNotFoundError:
    print("Hinweis: data/sessions.csv wurde nicht gefunden. Lege sie ggf. an.")

# Lösungshinweis:
# mit csv.DictReader arbeiten und FileNotFoundError abfangen.
````

## B6a – While-Schleife: Menü für Mediennutzung

Erstelle ein kleines Textmenü mit einer `while`-Schleife:

- Solange der/die Nutzer:in nicht `"3"` eingibt, soll das Menü erneut angezeigt werden.
- Optionen:
  - `1) Minuten für heute eingeben`
  - `2) Gesamtminuten anzeigen`
  - `3) Beenden`
- Bei Option 1 wird per `input()` eine Zahl abgefragt und zu einer Variable `total_minutes` addiert.
- Bei Option 2 wird `total_minutes` ausgegeben.

Nutze dazu eine `while`-Schleife und eine `if` / `elif` / `else`-Struktur.

````python
# TODO: B6a – While-Menü für Medienminuten
total_minutes = 0
choice = "0"

while choice != "3":
    print("\n--- Medienmenü ---")
    print("1) Minuten für heute eingeben")
    print("2) Gesamtminuten anzeigen")
    print("3) Beenden")
    choice = input("Auswahl: ")

    if choice == "1":
        text = input("Wie viele Minuten heute? ")
        try:
            minutes = int(text)
            if minutes < 0:
                print("Bitte keine negativen Minuten eingeben.")
            else:
                total_minutes += minutes
        except ValueError:
            print("Das war keine gültige Zahl.")
    elif choice == "2":
        print("Bisherige Gesamtminuten:", total_minutes)
    elif choice == "3":
        print("Programm wird beendet.")
    else:
        print("Ungültige Auswahl.")

# Hinweis: typische Kombination aus while-Schleife, input und if/elif.
````

## B7 – Sichere Integer-Eingabe

Schreibe eine Funktion `safe_int_input(prompt)`, die so lange nachfragt,
bis eine gültige ganze Zahl eingegeben wurde, und diese zurückgibt.

Nutze `try` / `except ValueError`.

````python
def safe_int_input(prompt):
    while True:
        text = input(prompt)
        try:
            return int(text)
        except ValueError:
            print("Bitte eine gültige ganze Zahl eingeben.")

# Beispiel (kannst du testen):
# value = safe_int_input("Ganze Zahl: ")
# print("Du hast eingegeben:", value)

# Lösungshinweis:
# while True + try/except.
````

## B8 – Robustes Profil-Reading

Gegeben ist ein Profil-Dictionary, bei dem nicht garantiert ist, dass alle Schlüssel existieren:

```python
profile = {"name": "Alex", "favorite_medium": "Musik"}
```

Schreibe eine Funktion `describe_profile(profile)`, die einen Text zurückgibt, z. B.:

`"Alex mag Musik und verbringt 0 Minuten pro Woche damit."`

Wenn `"minutes_per_week"` fehlt, soll standardmäßig 0 angenommen werden (kein Fehler!).

````python
def describe_profile(profile):
    name = profile.get("name", "(unbekannt)")
    medium = profile.get("favorite_medium", "Medien")
    minutes = profile.get("minutes_per_week", 0)
    return f"{name} mag {medium} und verbringt {minutes} Minuten pro Woche damit."

profile1 = {"name": "Alex", "favorite_medium": "Musik"}
profile2 = {"name": "Sam", "favorite_medium": "Film", "minutes_per_week": 300}

print(describe_profile(profile1))
print(describe_profile(profile2))

# Lösungshinweis:
# dict.get(key, default) verwenden.
````

## B10 – Mini-Projekt: Media Engagement Checker

Baue ein kleines Textprogramm, das:

1. nach Name, Medium und Minuten fragt (mit **sicherer Minuten-Eingabe**, z. B. `safe_int_input` aus B7)
2. einen Score berechnet (z. B. Minuten \* 0.5, maximal 100)
3. den Score mit `classify_score` aus B2 einstuft
4. am Ende einen Satz ausgibt, z. B.:

`"Alex nutzt Musik 120 Minuten/Woche – Score 60.0 (mittel)."`

Du kannst dazu Funktionen aus früheren Aufgaben **wiederverwenden**.

````python
# TODO: Mini-Projekt – Media Engagement Checker
# Du kannst z. B. safe_int_input() und classify_score() wiederverwenden.

def media_score(minutes):
    score = minutes * 0.5
    if score > 100:
        score = 100
    return score

# Beispiel (interaktive Variante kannst du selbst ergänzen):
name = "Alex"
medium = "Musik"
minutes = 120

score = media_score(minutes)

def classify_score(score):
    if score < 25:
        return "sehr wenig"
    elif score < 50:
        return "wenig"
    elif score < 75:
        return "mittel"
    else:
        return "viel"

category = classify_score(score)
print(f"{name} nutzt {medium} {minutes} Minuten/Woche – Score {score} ({category}).")

# Lösungshinweis:
# Teile das Problem in kleine Funktionen: score-Berechnung, Klassifikation, Ausgabe.
````
