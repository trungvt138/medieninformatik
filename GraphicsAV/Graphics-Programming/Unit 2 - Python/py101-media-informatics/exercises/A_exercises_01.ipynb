## A4 – Durchschnitt berechnen
Gegeben:
`minutes = [10, 20, 30, 25]`

```python
minutes = [10, 20, 30, 25]

total = sum(minutes)
avg = total / len(minutes)

print("Gesamt:", total)
print("Durchschnitt:", avg)

# Lösungshinweis: sum() und len() nutzen.
```

## A5 – Längsten Song ohne `max()` finden
Gegeben:
`songs = [180, 220, 210, 95, 300]`

```python
songs = [180, 220, 210, 95, 300]

longest = songs[0]
for s in songs:
    if s > longest:
        longest = s

print("Längster Song:", longest)

# Lösungshinweis: Mit Vergleich in Schleife.
```

## A7 – Minuten pro Medium addieren
Gegeben ist eine Liste von Dictionaries.

```python
sessions = [
    {"medium": "Musik", "minutes": 30},
    {"medium": "Film", "minutes": 90},
    {"medium": "Musik", "minutes": 20},
]

totals = {}
for s in sessions:
    m = s["medium"]
    minutes = s["minutes"]
    totals[m] = totals.get(m, 0) + minutes

print(totals)

# Lösungshinweis: dict.get() nutzen.
```

## A9 – Sekunden in Minuten/Sekunden umwandeln

```python
def to_min_sec(seconds):
    minutes = seconds // 60
    secs = seconds % 60
    return minutes, secs

print(to_min_sec(197))

# Lösungshinweis: Ganzzahldivision und Rest.
```

## A11 – While-Schleife: Zähle bis 10

Schreibe eine `while`-Schleife, die die Zahlen **1 bis 10** ausgibt.

Hinweis: Du brauchst eine Variable, die du in jedem Schleifendurchlauf erhöhst.

```python
# TODO: Zähle mit while von 1 bis 10
num = 1
while num <= 10:
    print(num)
    num += 1

# Lösungshinweis:
# Starte bei 1 und erhöhe in jedem Schritt um 1.
```

## A12 – Mini-Menü mit while

Erstelle ein kleines Menü, das so lange läuft, bis der/die Nutzer:in `"q"` eingibt.

Menü:
- `1` → „Hallo!“ ausgeben
- `2` → „Python macht Spaß!“ ausgeben
- `q` → Programm beenden

Nutze dazu eine `while`-Schleife und `input()`.

```python
# TODO: Mini-Menü mit while-Schleife
choice = ""
while choice != "q":
    print("\n--- Menü ---")
    print("1) Hallo sagen")
    print("2) Python macht Spaß")
    print("q) Beenden")
    choice = input("Auswahl: ")

    if choice == "1":
        print("Hallo!")
    elif choice == "2":
        print("Python macht Spaß!")
    elif choice == "q":
        print("Programm beendet.")
    else:
        print("Ungültige Eingabe.")
```
