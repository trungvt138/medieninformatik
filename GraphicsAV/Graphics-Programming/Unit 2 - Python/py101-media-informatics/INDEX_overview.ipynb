## 6. Arbeiten in GitHub Codespaces / VS Code

Kurzfassung:

1. Repo öffnen (lokal in VS Code oder als GitHub Codespace).  
2. Sicherstellen, dass der Python-Interpreter / die Umgebung richtig gewählt ist.  
3. Im Explorer links ein Notebook (`.ipynb`) anklicken → es öffnet sich im Notebook-Editor.  
4. Oben rechts Kernel auswählen (z. B. den vordefinierten Projekt-Interpreter).  
5. Zelle(n) ausführen und bearbeiten.

➡️ Details & Screenshots:  
[docs/using_codespaces.md](docs/using_codespaces.md)

Wenn etwas mit der Umgebung nicht funktioniert, siehe den Abschnitt *Fehlersuche* in `using_codespaces.md` oder frage im Kurs nach Hilfe.
