## 1. Einfache `if`-Anweisung

Eine `if`-Anweisung prüft eine Bedingung. Ist die Bedingung `True`, wird der eingerückte Block ausgeführt.

```python
x = 10

if x > 5:
    print("x ist größer als 5")
```

- Nach `if` steht eine Bedingung.
- Am Zeilenende steht ein Doppelpunkt `:`.
- Der zugehörige Block ist eingerückt (z. B. 4 Leerzeichen).

````python
x = 10

if x > 5:
    print("x ist größer als 5")

print("Diese Zeile wird immer ausgeführt.")
````

## 3. `if` – `elif` – `else`: Mehrere Fälle

Mit `elif` ("else if") kannst du mehrere Bedingungen der Reihe nach prüfen.

```python
punkte = 83

if punkte >= 90:
    note = 1
elif punkte >= 80:
    note = 2
elif punkte >= 70:
    note = 3
else:
    note = 4
```

````python
punkte = 83

if punkte >= 90:
    note = 1
elif punkte >= 80:
    note = 2
elif punkte >= 70:
    note = 3
else:
    note = 4

print("Punkte:", punkte, "→ Note:", note)
````

## 4. Bedingungen kombinieren: `and`, `or`, `not`

Du kannst mehrere Bedingungen kombinieren.

Beispiele:

```python
if alter >= 18 and hat_ausweis:
    print("Du darfst rein.")

if medium == "Film" or medium == "Serie":
    print("Bewegtbild")

if not ist_student:
    print("Kein Studienstatus")
```

````python
alter = 20
hat_ausweis = True
medium = "Film"
ist_student = False

if alter >= 18 and hat_ausweis:
    print("Du darfst in den 18+ Bereich.")

if medium == "Film" or medium == "Serie":
    print("Du schaust Bewegtbild.")

if not ist_student:
    print("Du scheinst kein*e Student*in zu sein.")
````

## 5. Verschachtelte `if`-Anweisungen

Manchmal möchtest du **innen drin** noch eine weitere Entscheidung treffen. Das nennt man _verschachtelte_ `if`-Blöcke.

```python
alter = 19
ist_premium = True

if alter >= 18:
    if ist_premium:
        print("Voller 18+ Premium-Zugriff")
    else:
        print("18+, aber ohne Premium")
else:
    print("Unter 18 – kein Zugriff auf 18+ Inhalte")
```

````python
alter = 19
ist_premium = True

if alter >= 18:
    if ist_premium:
        print("Voller 18+ Premium-Zugriff")
    else:
        print("18+, aber ohne Premium")
else:
    print("Unter 18 – kein Zugriff auf 18+ Inhalte")
````

## 6. While-Schleifen – Wiederholungen mit Bedingungen

Bisher hast du mit `if`-Anweisungen **Entscheidungen** getroffen und einen Codeblock ausgeführt wenn eine Bedingung **wahr** ist.
Mit einer `while`-Schleife kannst du einen Codeblock so lange wiederholen,
wie eine Bedingung **wahr** ist.

Grundform:

```python
count = 1
while count <= 3:
    print("Durchlauf", count)
    count += 1
```

Wichtig:
- es gibt eine **Startvariable** (hier `count`)
- die Bedingung (`count <= 3`) entscheidet, ob weiter gemacht wird
- innerhalb der Schleife wird die Variable **verändert**, sonst läuft die Schleife unendlich


````python
# Beispiel: einfache while-Schleife
count = 1
while count <= 3:
    print("Durchlauf", count)
    count += 1
````
