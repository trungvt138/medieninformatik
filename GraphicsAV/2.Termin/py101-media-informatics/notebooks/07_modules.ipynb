## 1. Was ist ein Modul?
In Python ist **jede `.py`-Datei ein Modul**.
Beispiele für eingebaute/bekannte Module:
- `math` – Mathefunktionen
- `random` – Zufallszahlen
- `datetime` – Datum/Zeit
Du kannst sie so verwenden:```python
import math
print(math.sqrt(16))
```Oder mit einer Kurzform:```python
import math as m
print(m.sqrt(16))
```Oder du importierst nur bestimmte Funktionen:```python
from math import sqrt
print(sqrt(16))
```

`````python
import math
print("Wurzel aus 25:", math.sqrt(25))
from random import randint
print("Zufallszahl zwischen 1 und 10:", randint(1, 10))
`````

## 2. Eigene Module erstellen
Angenommen, du legst im Projekt eine Datei an:```text
python-selfstudy-medieninformatik/
└─ src/
   └─ media_utils.py
```In `media_utils.py` steht z. B.:```python
def seconds_to_min_sec(total_seconds):
    minutes = total_seconds // 60
    seconds = total_seconds % 60
    return minutes, seconds
```Dann kannst du in einem anderen Skript oder Notebook schreiben:```python
from src import media_utils
m, s = media_utils.seconds_to_min_sec(197)
print(m, s)
```oder:```python
from src.media_utils import seconds_to_min_sec
m, s = seconds_to_min_sec(197)
```

## 3. Varianten von `import`
Du kannst Module auf verschiedene Arten importieren. Hier ein Überblick:### 3.1 Ganzes Modul```python
import media_utils
media_utils.seconds_to_min_sec(200)
```- Vorteil: Du siehst immer, aus welchem Modul eine Funktion stammt.
- Nachteil: Längere Schreibweise.### 3.2 Einzelne Namen aus einem Modul```python
from media_utils import seconds_to_min_sec
seconds_to_min_sec(200)
```- Vorteil: Kürzer im Code.
- Nachteil: Kann unübersichtlich werden, wenn viele Funktionen aus vielen Modulen kommen.### 3.3 Aliasnamen vergeben```python
import media_utils as mu
mu.seconds_to_min_sec(200)
```- Praktisch, wenn Modulnamen lang sind oder du eine bestimmte Schreibweise im Projekt nutzen willst.

## 4. Ordner als Pakete nutzen
Wenn du beginnst, deinen Code zu strukturieren, arbeitest du oft mit Unterordnern:```text
python-selfstudy-medieninformatik/
└─ src/
   ├─ __init__.py
   ├─ media_utils.py
   └─ analysis.py
```- Die Datei `__init__.py` (kann leer sein) markiert `src/` als **Python-Paket**.
- Du kannst dann z. B. schreiben:```python
from src.media_utils import seconds_to_min_sec
```In vielen modernen Projekten (und mit Tools wie VS Code & GitHub Codespaces) funktioniert der Import aus `src/` oft auch ohne explizites `__init__.py`, aber es ist gut, das Prinzip zu kennen.

## 5. Notebooks & Module gemeinsam nutzen
Typischer Workflow im Kurskontext:- **Notebooks** (`.ipynb`) für:
  - Erklärungen
  - interaktive Experimente
  - kleine Tests

- **Module** (`.py`) für:
  - wiederverwendbare Funktionen
  - sauberen, getesteten Code
Beispiel-Struktur:```text
python-selfstudy-medieninformatik/
├─ notebooks/
│  ├─ 03_functions.ipynb
│  ├─ 04_data_structures.ipynb
│  └─ 07_modules.ipynb
└─ src/
   ├─ __init__.py
   └─ media_utils.py
```Im Notebook kannst du dann einfach:```python
from src.media_utils import seconds_to_min_sec
```### Achtung: Module neu ladenWenn du ein Modul geändert hast, während dein Notebook noch läuft, musst du es evtl. **neu laden**:```python
import importlib
import src.media_utils as mu
# später, nach Änderung an media_utils.py
importlib.reload(mu)
```Das ist optional, aber praktisch, wenn du intensiv Notebook + Module mischst.

## 6. Beispiel: Kleine Medien-Projektstruktur
Stell dir vor, du baust ein kleines Tool, das Medien-Sessions analysiert.Projektstruktur:```text
media-project/
├─ src/
│  ├─ __init__.py
│  ├─ data_models.py
│  └─ analysis.py
└─ main.py
```### `src/data_models.py````python
def create_session(user, medium, minutes):
    if minutes < 0:
        raise ValueError("minutes must be >= 0")
    return {"user": user, "medium": medium, "minutes": minutes}
```### `src/analysis.py````python
def total_minutes_per_user(sessions):
    totals = {}
    for s in sessions:
        name = s["user"]
        totals[name] = totals.get(name, 0) + s["minutes"]
    return totals
```### `main.py````python
from src.data_models import create_session
from src.analysis import total_minutes_per_user

sessions = [
    create_session("Alex", "Musik", 30),
    create_session("Alex", "Film", 60),
    create_session("Sam", "Game", 45),
]

totals = total_minutes_per_user(sessions)
print(totals)
```So trennst du **Struktur/Logik** (in `src/`) von einem einfachen **Einstiegspunkt** (`main.py`).

```### `media_time.py` (Beispiel)```python
def minutes_to_hours(minutes):
    return minutes / 60

def format_hours(minutes):
    hours = minutes_to_hours(minutes)
    return f"{hours:.2f} Stunden"
```### `media_profile.py` (Beispiel)```python
def create_profile(name, favorite_medium, minutes_per_week):
    return {
        "name": name,
        "favorite_medium": favorite_medium,
        "minutes_per_week": minutes_per_week,
    }
```### Im Notebook oder in `main.py````python
from src.media_time import format_hours
from src.media_profile import create_profile

p = create_profile("Alex", "Musik", 600)
print(p["name"], "hört", format_hours(p["minutes_per_week"]))
