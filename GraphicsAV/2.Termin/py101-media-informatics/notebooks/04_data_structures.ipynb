## 1. Listen (`list`)

Listen sind **geordnete**, veränderbare Sammlungen von Elementen.

```python
numbers = [1, 2, 3]
names = ["Alex", "Sam", "Riley"]
mixed = [1, "Hallo", 3.14]
```

Du kannst über den **Index** zugreifen (beginnend bei 0):

```python
names[0]  # "Alex"
```

````python
numbers = [1, 2, 3, 4]
names = ["Alex", "Sam", "Riley"]

print("numbers:", numbers)
print("names:", names)
print("Erstes Element von names:", names[0])
print("Länge von names:", len(names))
````

### Wichtige Listen-Operationen

- `append(x)` – ans Ende anhängen
- `insert(i, x)` – an Position `i` einfügen
- `remove(x)` – erstes Vorkommen von `x` entfernen
- `pop(i)` – Element an Position `i` entfernen und zurückgeben
- `in` – prüfen, ob ein Element enthalten ist


````python
playlist = ["Song A", "Song B", "Song C"]
print("Start:", playlist)

playlist.append("Song D")
print("Nach append:", playlist)

playlist.insert(1, "Intro")
print("Nach insert an Position 1:", playlist)

playlist.remove("Song B")
print("Nach remove('Song B'):", playlist)

last = playlist.pop()
print("Nach pop():", playlist, "(entfernt:", last, ")")

print("Enthält 'Song A'?:", "Song A" in playlist)
````

## 2. Tupel (`tuple`)

Tupel sind **geordnete, aber unveränderliche** Sammlungen.

```python
position = (100, 200)
color = (255, 128, 0)
```

Du kannst zugreifen wie bei Listen, aber nicht verändern:

```python
position[0]  # 100
```

````python
screen_size = (1920, 1080)
print("Breite:", screen_size[0])
print("Höhe:", screen_size[1])

# screen_size[0] = 1280  # würde einen Fehler auslösen (unveränderlich)
````

## 3. Dictionaries (`dict`)

Dictionaries speichern **Schlüssel-Wert-Paare**.

```python
user = {
    "name": "Alex",
    "age": 21,
    "study_program": "Medieninformatik"
}
```

Zugriff über den **Schlüssel**:

```python
user["name"]  # "Alex"
```

````python
user = {
    "name": "Alex",
    "age": 21,
    "study_program": "Medieninformatik"
}

print("Name:", user["name"])
print("Alter:", user["age"])
print("Studiengang:", user["study_program"])
````

### Einträge hinzufügen, ändern, löschen

- Neuen Schlüssel hinzufügen:
  ```python
  user["favorite_medium"] = "Film"
  ```
- Wert ändern:
  ```python
  user["age"] = 22
  ```
- Eintrag löschen:
  ```python
  del user["age"]
  ```
- Prüfen, ob ein Schlüssel existiert:
  ```python
  "name" in user
  ```

````python
print("Start:", user)

user["favorite_medium"] = "Film"
print("Nach hinzufügen von favorite_medium:", user)

user["age"] = 22
print("Nach Änderung von age:", user)

print("Hat Schlüssel 'name'?:", "name" in user)

del user["study_program"]
print("Nach del study_program:", user)
````

### Über Dictionaries iterieren

Du kannst über Schlüssel, Werte oder beides iterieren:

```python
for key in user.keys():
    ...

for value in user.values():
    ...

for key, value in user.items():
    ...
```

````python
print("Schlüssel:")
for key in user.keys():
    print(" -", key)

print("Werte:")
for value in user.values():
    print(" -", value)

print("Schlüssel und Werte:")
for key, value in user.items():
    print(f" - {key}: {value}")
````

## 4. Sets (`set`)

Sets sind **ungeordnete Mengen ohne Duplikate**.

```python
tags = {"music", "film", "game"}
```

Du kannst Elemente hinzufügen oder entfernen, aber es gibt keine festen Positionen (keine Indizes).

````python
tags = {"music", "film", "game"}
print("Start:", tags)

tags.add("podcast")
print("Nach add('podcast'):", tags)

tags.add("music")  # Duplikat – Set bleibt gleich
print("Nach add('music') (Duplikat):", tags)

tags.remove("film")
print("Nach remove('film'):", tags)

print("Enthält 'game'?:", "game" in tags)
````

### Mengenoperationen

Sets unterstützen typische Mengenoperationen:

- Vereinigung (`|`)
- Schnittmenge (`&`)
- Differenz (`-`)

Beispiel:

```python
user1_tags = {"music", "film"}
user2_tags = {"music", "game"}

gemeinsam = user1_tags & user2_tags       # Schnittmenge
alle = user1_tags | user2_tags            # Vereinigung
nur_user1 = user1_tags - user2_tags       # Differenz
```

````python
user1_tags = {"music", "film"}
user2_tags = {"music", "game"}

gemeinsam = user1_tags & user2_tags
alle = user1_tags | user2_tags
nur_user1 = user1_tags - user2_tags

print("User1-Tags:", user1_tags)
print("User2-Tags:", user2_tags)
print("Gemeinsam:", gemeinsam)
print("Alle:", alle)
print("Nur User1:", nur_user1)
````

## 5. Strukturen kombinieren

## 6. Abschlussaufgabe – Einfaches Medien-Tracking

Baue ein kleines Datenmodell für ein Medien-Tracking-System.

Vorgabe:

- Du verwaltest eine Liste `sessions`, in der jede Sitzung ein Dictionary ist mit:
  - `"user"` – Name der Person
  - `"medium"` – z. B. `"Film"`, `"Musik"`, `"Game"`, `"Podcast"`
  - `"minutes"` – Dauer der Sitzung

Beispiel:

```python
sessions = [
    {"user": "Alex", "medium": "Musik", "minutes": 30},
    {"user": "Alex", "medium": "Game", "minutes": 45},
    {"user": "Sam", "medium": "Film", "minutes": 120},
]
```

Aufgaben:

1. Erzeuge eine Beispiel-Liste `sessions` mit mindestens 5 Einträgen.
2. Berechne die **Gesamtminuten pro Nutzer** (z. B. mit einem Dictionary `total_per_user`).
3. Berechne optional auch die Minuten pro Medium (z. B. `total_per_medium`).
4. Gib die Ergebnisse lesbar aus.

Tipp: du kannst z. B. so vorgehen:

```python
total_per_user = {}
for session in sessions:
    name = session["user"]
    minutes = session["minutes"]
    if name not in total_per_user:
        total_per_user[name] = 0
    total_per_user[name] += minutes
```

````python
# TODO: Abschlussaufgabe – Medien-Tracking
sessions = [
    {"user": "Alex", "medium": "Musik", "minutes": 30},
    {"user": "Alex", "medium": "Game", "minutes": 45},
    {"user": "Sam", "medium": "Film", "minutes": 120},
    {"user": "Meylan", "medium": "Film", "minutes": 145},
    {"user": "Meylan", "medium": "Game", "minutes": 65}
]

total_per_user = {}
for session in sessions:
    name = session["user"]
    minutes = session["minutes"]
    if name not in total_per_user:
        total_per_user[name] = 0
    total_per_user[name] += minutes

total_per_medium = {}
for session in sessions:
    medium = session["medium"]
    minutes = session["minutes"]
    if medium not in total_per_medium:
        total_per_medium[medium] = 0
    total_per_medium[medium] += minutes

print(total_per_user)
print(total_per_medium)
````
