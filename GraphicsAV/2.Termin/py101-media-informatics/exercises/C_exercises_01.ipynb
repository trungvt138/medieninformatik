## C1 – Zusammenfassung mit List Comprehension

Gegeben ist eine Liste von Sessions:

```python
sessions = [
    {"user": "Alex", "medium": "Musik", "minutes": 30},
    {"user": "Sam", "medium": "Film", "minutes": 90},
    {"user": "Alex", "medium": "Game", "minutes": 45},

```

Erzeuge mit einer **List Comprehension** eine neue Liste von Strings, z. B.:

```python

```

## C3 – Durchschnittsminuten pro Nutzer

    Schreibe eine Funktion `average_minutes_per_user(sessions)`, die ein Dictionary zurückgibt:

```python
{"Alex": 37.5, "Sam": 90.0}
```

Nutze dazu zuerst eine Aggregation der Gesamtminuten und Anzahl der Sessions pro Nutzer
und berechne dann den Durchschnitt.

````python
def average_minutes_per_user(sessions):
    totals = {}
    counts = {}
    for s in sessions:
        name = s["user"]
        mins = s["minutes"]
        totals[name] = totals.get(name, 0) + mins
        counts[name] = counts.get(name, 0) + 1
    averages = {name: totals[name] / counts[name] for name in totals}
    return averages

print(average_minutes_per_user(sessions))

# Hinweis: Du kannst das averages-Dict mit einer Dict Comprehension bauen.
````

## C4 – Verschachtelte Struktur flatten

Gegeben:

```python
user_sessions = {
    "Alex": [
        {"medium": "Musik", "minutes": 30},
        {"medium": "Game", "minutes": 45},
    ],
    "Sam": [
        {"medium": "Film", "minutes": 90},
    ],
}
```

Erzeuge eine **flache Liste** aller Sessions mit explizitem `"user"`-Feld, z. B.:

```python
[{"user": "Alex", "medium": "Musik", "minutes": 30}, ...]
```

````python
user_sessions = {
    "Alex": [
        {"medium": "Musik", "minutes": 30},
        {"medium": "Game", "minutes": 45},
    ],
    "Sam": [
        {"medium": "Film", "minutes": 90},
    ],
}

flat_sessions = [
    {"user": user, "medium": s["medium"], "minutes": s["minutes"]}
    for user, sess_list in user_sessions.items()
    for s in sess_list
]

for s in flat_sessions:
    print(s)

# Hinweis: verschachtelte for-Schleifen in einer List Comprehension.
````

## C5 – Ranking nach Score

Nimm `flat_sessions` aus Aufgabe C4 und berechne einen **Score pro Nutzer**:

- Score = Gesamtminuten über alle Sessions (oder gerne komplexer, z. B. Gewichtung nach Medium)

    Erzeuge dann eine **sortierte Liste** von `(user, score)`-Tupeln, absteigend nach Score.

Nutze `sorted(..., key=..., reverse=True)`.

````python
def total_minutes_per_user(sessions):
    totals = {}
    for s in sessions:
        name = s["user"]
        mins = s["minutes"]
        totals[name] = totals.get(name, 0) + mins
    return totals

totals = total_minutes_per_user(flat_sessions)
ranking = sorted(totals.items(), key=lambda item: item[1], reverse=True)

print("Ranking (user, total_minutes):")
for user, mins in ranking:
    print(f"- {user}: {mins} Minuten")

# Hinweis: totals.items() gibt (key, value)-Tupel.
````

## C6 – Funkionen höherer Ordnung

Schreibe eine Funktion `compute_scores(sessions, scoring_fn)`, die:

- eine Liste von Sessions nimmt
- eine Funktion `scoring_fn(user, sessions_for_user)` erwartet
- ein Dictionary `user → score` zurückgibt

Definiere dann **zwei verschiedene** `scoring_fn`-Varianten:

1. Score = Gesamtminuten
2. Score = Durchschnittsminuten pro Session

````python
def group_sessions_by_user(sessions):
    groups = {}
    for s in sessions:
        name = s["user"]
        groups.setdefault(name, []).append(s)
    return groups

def compute_scores(sessions, scoring_fn):
    groups = group_sessions_by_user(sessions)
    scores = {}
    for user, sess_list in groups.items():
        scores[user] = scoring_fn(user, sess_list)
    return scores

def score_total_minutes(user, sess_list):
    return sum(s["minutes"] for s in sess_list)

def score_average_minutes(user, sess_list):
    if not sess_list:
        return 0
    return sum(s["minutes"] for s in sess_list) / len(sess_list)

print("Scores (total):", compute_scores(flat_sessions, score_total_minutes))
print("Scores (average):", compute_scores(flat_sessions, score_average_minutes))

# Hinweis: Funktionen können wie Daten übergeben werden.
````

## C7 – Mini-Testfunktionen mit `assert`

Schreibe kleine Testfunktionen für `format_duration` (aus Level B) oder eine deiner neuen Funktionen, z. B.:

```python
def test_format_duration():
    assert format_duration(60) == "1:00"
    assert format_duration(197) == "3:17"
```

Rufe deine Testfunktionen am Ende auf. Wenn ein `assert` fehlschlägt, bekommst du eine Fehlermeldung – sonst bleibt es still.

````python
def format_duration(seconds):
    minutes = seconds // 60
    rest = seconds % 60
    return f"{minutes}:{rest:02d}"

def test_format_duration():
    assert format_duration(60) == "1:00"
    assert format_duration(197) == "3:17"
    assert format_duration(0) == "0:00"

test_format_duration()
print("Alle Tests für format_duration bestanden.")

# Hinweis: assert ist eine einfache Möglichkeit, Erwartungen zu prüfen.
````

## C8 – Datenvalidierung mit Exceptions

Schreibe eine Funktion `validate_session(session)`, die prüft:

- `"user"` existiert und ist ein nicht-leerer String
- `"minutes"` existiert und ist ≥ 0

Wenn etwas nicht passt, wirf einen `ValueError` mit einer sinnvollen Nachricht.

Schreibe dann eine Funktion `add_session_safe(sessions, session)`, die `validate_session` aufruft und die Session nur anhängt, wenn sie gültig ist. Im Fehlerfall soll eine Warnung ausgegeben werden, aber das Programm nicht abbrechen.

````python
def validate_session(session):
    if "user" not in session or not isinstance(session["user"], str) or not session["user"].strip():
        raise ValueError("Session muss ein nicht-leeres 'user'-Feld haben.")
    if "minutes" not in session or session["minutes"] < 0:
        raise ValueError("Session muss 'minutes' ≥ 0 haben.")

def add_session_safe(sessions, session):
    try:
        validate_session(session)
        sessions.append(session)
    except ValueError as e:
        print("Warnung: Session verworfen:", e)

sessions_c8 = []
add_session_safe(sessions_c8, {"user": "Alex", "minutes": 30})
add_session_safe(sessions_c8, {"user": " ", "minutes": 10})  # ungültig
add_session_safe(sessions_c8, {"user": "Sam", "minutes": -5})  # ungültig
print("Gespeicherte Sessions:", sessions_c8)

# Hinweis: Diese Art Validierung brauchst du in „echten“ Projekten ständig.
````
